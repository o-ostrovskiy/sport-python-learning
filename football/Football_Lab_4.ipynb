## Part 1. Your first list 📋
**Guess first:** what will this box show?

```python
my_board = ["Justin Jefferson", "Saquon Barkley", "Josh Allen"]
print(my_board)
```

That's a **list**: one box that holds many things, **in order**.
(Python shows the names with single quotes `'Justin Jefferson'`. Single or double quotes mean the same thing.)

| piece | what it does |
|---|---|
| `[` and `]` | square brackets: where the list **starts** and **ends** |
| `"Justin Jefferson"` | each name is in quotes, because names are text (Lab 2) |
| `,` | a comma between players, so Python knows where one ends and the next begins |
| `my_board =` | puts the whole list into one box called `my_board` |

**Seen this before?** In Lab 2 you wrote `stats[["player_display_name", "receiving_yards"]]`.
The inside brackets `["player_display_name", "receiving_yards"]` were a list too: a list of column names!

**Why the commas?** Watch what happens without one. **Guess first:** how many players are on this board?

```python
no_comma = ["Justin Jefferson" "Saquon Barkley"]
print(no_comma)
print(len(no_comma))
```

No red error, but it's **wrong**: Python glued the two names into one player, `Justin JeffersonSaquon Barkley`,
and `len( )` (Lab 2: "how many?") says there's only **1** player. The comma is what keeps players apart.

## Part 2. Pick number... 0?! 🤯
A bigger board. A list this long doesn't fit on one line, so we write **one player per line**.
That's allowed: Python keeps reading until it finds the closing bracket `]`.
(Ja'Marr Chase has a `'` in his name. That's fine inside double quotes `" "`.)

**Guess first:** `my_board[1]` asks for one player from the board. Which one will it show?

```python
my_board = [
    "Justin Jefferson",
    "Saquon Barkley",
    "Josh Allen",
    "Ja'Marr Chase",
    "Bijan Robinson",
]
print(my_board[1])
```

Did you guess Justin Jefferson? Almost everyone does. But Python says **Saquon Barkley**!

**Python counts from 0, not 1.** The number in the brackets means "how many steps from the top".
Justin Jefferson is at the very top: **0 steps**. Saquon is 1 step down.

| pick on your board | 1st | 2nd | 3rd | 4th | 5th |
|---|---|---|---|---|---|
| the number Python uses | `0` | `1` | `2` | `3` | `4` |

`my_board[1]` means: "from `my_board`, give me the one at spot 1".
It's the same square-bracket idea as `stats["receiving_yards"]` in Lab 2: brackets after a name **pick something out of it**.

```python
print("Pick 1:", my_board[0])
print("Pick 2:", my_board[1])
print("Pick 3:", my_board[2])
```

There are 5 players, so the spots are 0, 1, 2, 3, 4. What if we ask for spot **5**?
Run this box. It shows an error **on purpose**:

```python
print(my_board[5])
```

The last line says `IndexError: list index out of range`.
**Index** is Python's word for the spot number. "Out of range" means there's no spot 5 on this board:
5 players only go up to spot 4. **The last spot is always one less than `len( )`.**

## Part 3. Draft day 📞
### `.append( )`: add a player to the end
You found a new player you like. **Guess first:** where on the board will he go?

```python
my_board = ["Justin Jefferson", "Saquon Barkley", "Josh Allen"]
my_board.append("Lamar Jackson")
print(my_board)
print("Players on my board:", len(my_board))
```

- **`append`** means "add to the end". Lamar goes to the bottom of the board, and the board grows from 3 to 4.
- **The dot** `.` you know from Lab 2 (`.max()`): it means "do this to that".
  `my_board.append("Lamar Jackson")` = "to `my_board`, add Lamar Jackson at the end".
- The `append` line doesn't show anything by itself. It changes the list quietly. `print` shows it.

### `in`: is he on my board?
**Guess first:** what will each line say?

```python
my_board = ["Justin Jefferson", "Saquon Barkley", "Josh Allen"]

print("Josh Allen" in my_board)
print("Patrick Mahomes" in my_board)
print("josh allen" in my_board)
```

- `in` asks a yes-or-no question: "is this **in** the list?" Like the symbols in Lab 3, it answers `True` or `False`.
- `"Josh Allen" in my_board` → `True`. `"Patrick Mahomes" in my_board` → `False`, he's not on the board.
- `"josh allen"` → `False`! To Python, a small `j` and a big `J` are different letters (Lab 2).
  `in` only says `True` when the name is spelled **exactly** the same.

### 🏈 Your turn: the draft-day machine
This box puts it all together with `if` and `else` from Lab 3:
**if the player is already on the board, say so; otherwise, add him.**

One word is missing in the `if` line: the one that asks "is it inside the list?". It's 2 letters.
And one is missing under `else`: the one that adds a player to the end. It's 6 letters.

🔎 **Google it:** `python check if item in list` and `python add item to end of list`

```python
my_board = ["Justin Jefferson", "Saquon Barkley", "Josh Allen"]
player = "Patrick Mahomes"   # change to any player

if player ___ my_board:
    print(player, "is already on my board ✅")
else:
    my_board.___(player)
    print("Added", player, "to the end of my board 📋")

print(my_board)
```

- Under `else` there are **two** indented lines. Both belong to `else`, so both happen only when the player is **not** on the board.
- The last `print(my_board)` is **not** indented, so it runs every time.

Run it twice: once with `player = "Patrick Mahomes"`, then with `player = "Josh Allen"`.
The second time, nobody gets added: the machine sees he's already there. That's how real teams avoid picking the same player twice!

## Part 4. The real top 5 📊
Real teams build their boards from stats. Let's do it too.

This is a **new file**: in Labs 2 and 3, one row was **one game**. Here one row is **one whole season** of one player,
already added up for you. It loads in a few seconds.

```python
import pandas as pd

seasons = pd.read_csv(
    "https://github.com/nflverse/nflverse-data/releases/download/player_stats/player_stats_season.csv",
    low_memory=False,
)
seasons = seasons[seasons["season_type"] == "REG"]
newest = seasons["season"].max()
print("Ready!", len(seasons), "seasons loaded. The newest season is", newest)
```

`seasons["season"].max()` finds the biggest number in the `season` column (`.max()` from Lab 2), so `newest` is the newest season in the data.

Now three steps, each one you already know from Lab 2:

```python
last = seasons[seasons["season"] == newest]
best = last.sort_values("receiving_yards", ascending=False)
names = best["player_display_name"].head(5)
print(names)
```

| line | what it does |
|---|---|
| `last = seasons[seasons["season"] == newest]` | keep only the rows from the newest season |
| `best = last.sort_values("receiving_yards", ascending=False)` | put them in order, **most** receiving yards first |
| `names = best["player_display_name"].head(5)` | take the name column, just the top 5 |

The numbers on the left are the table's row labels, and the last line (`Name: ..., dtype: str`) is the table's
note about the column. Ignore them both.

`names` is still a piece of the table, not a list. One new command turns it into a real list: **`list( )`**.

```python
top5 = list(names)
print(top5)
```

Square brackets, commas, quotes: it's a list now! Ja'Marr Chase is shown with double quotes because his name has a `'` in it.

Now `top5` is a list, so everything from today works on it. **Guess first:**
who is the number 1 receiver? And is Justin Jefferson in the top 5?

```python
print("Number 1:", top5[0])
print("Is Justin Jefferson in the top 5?", "Justin Jefferson" in top5)
```
