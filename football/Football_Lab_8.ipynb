### `input( )` always gives text
Run the next box and type **4**. **Guess first:** what will `tds + tds` show?

### 🏈 Your turn: `int( )` makes it a number

first `input( )` asks and gets the text `"4"`, then `int( )` turns it into the number `4`.

## Part 3. Again, and again: `while` 🔁
In Lab 5 you met the `for` loop: it walks through a list and then stops.
A **`while`** loop is different: it keeps going **as long as** its question is True.
In English: "**while** I still have tries left, keep asking".

**Guess first:** which numbers will this print?

````python
tries = 0

while tries < 3:
    print("Try number", tries)
    tries = tries + 1

print("The loop is over. tries is now", tries)
````

It printed 0, 1, 2. Here's how Python read it:

| line | what it does |
|---|---|
| `tries = 0` | start counting at 0 |
| `while tries < 3:` | before **every** lap, ask: "is `tries` less than 3?" True → do the indented lines. False → stop |
| `print("Try number", tries)` | indented, so it's **inside** the loop |
| `tries = tries + 1` | also inside: add 1 to the count (the running total from Lab 5) |
| `print("The loop is over...` | not indented, so it runs **once**, after the loop |

When `tries` reached 3, the question `3 < 3` was False, so the loop stopped. Colon and indent work just like `if` and `for`.

**Guess first:** your team gains 3 yards on every play and needs 10 for a first down. How many plays will this loop run?

````python
yards = 0
plays = 0

while yards < 10:
    yards = yards + 3
    plays = plays + 1

print("First down after", plays, "plays, with", yards, "yards")
````

**4 plays.** After 3 plays there are only 9 yards, and 9 is still less than 10, so the loop goes one more lap: 12 yards.
A `for` loop needs to know the number of laps before it starts. A `while` loop finds out along the way.

### ⚠️ The endless loop
What if someone forgets the line `tries = tries + 1`?

```python
tries = 0

while tries < 3:
    print("Try number", tries)
```

`tries` stays 0 forever. 0 is always less than 3, so the question is always True and the loop **never stops**.
It would print `Try number 0` thousands of times. (Don't run this one.)
If it ever happens to you: press the **square stop button ⏹** on the left of the box.

**The rule:** every `while` needs a line inside it that changes the thing in its question.

### Two small new things: `!=` and `and`
**Guess first:** five answers, True or False?

````python
print("Josh Allen" == "Josh Allen")
print("Josh Allen" != "Josh Allen")
print("Josh Allen" != "Joe Burrow")

print(2 < 3 and 5 < 9)
print(2 < 3 and 5 < 1)
````

- **`!=`** asks "are these **not** the same?". It's the opposite of `==` (Lab 3). Think of it as an equals sign with a line through it: ≠.
- **`and`** goes between two questions. It says True only when **both** are True.
  `2 < 3 and 5 < 1` is False, because `5 < 1` is not true.

### 🏈 Your turn: a guessing loop
Now a real one: "**while** the guess is not the secret **and** there are tries left, ask again".

One symbol is missing: the one that means "less than", so you get 3 tries.
Fill it in, run it, and answer. Spell the name with capital letters!

🔎 **Google it:** `python while loop with counter`

````python
secret = "Saquon Barkley"
tries = 0
guess = ""

while guess != secret and tries ___ 3:
    guess = input("Who ran for 2,005 yards in 2024? ")
    tries = tries + 1

if guess == secret:
    print("Correct! You needed", tries, "tries.")
else:
    print("Out of tries. It was", secret)
````

- `guess = ""` is **empty text**: two quotes with nothing inside. The box `guess` has to exist before the `while` line asks about it.
- The loop can stop for **two** reasons: you got it right (`guess != secret` turns False), or you ran out of tries (`tries < 3` turns False).
- After the loop, `if` checks which one it was.

Run it again and get it wrong three times on purpose. See how the count stops the loop? That count is what keeps it from running forever.

## Part 4. Guess the Player 🏈
Time to build the real quiz. Python will pick a **secret player** from the real stats and give you clues.

First we need the players. Everything in this box you know from Lab 4:
load the seasons, keep the newest one, sort by receiving yards, take the top 10. It loads in a few seconds.

````python
import pandas as pd

seasons = pd.read_csv(
    "https://github.com/nflverse/nflverse-data/releases/download/player_stats/player_stats_season.csv",
    low_memory=False,
)
seasons = seasons[seasons["season_type"] == "REG"]
newest = seasons["season"].max()
last = seasons[seasons["season"] == newest]

top = last.sort_values("receiving_yards", ascending=False).head(10)
top = top[["player_display_name", "position", "recent_team", "receiving_yards"]]
top
````

The top 10 receivers of the newest season. Python will hide one of them.
Keep this table in sight: you'll need the spelling of the names.

### From a table to player cards
One new command: **`.to_dict("records")`** turns every row of the table into a **player card**: a dictionary, like in Lab 7.
**Guess first:** how many cards will there be?

````python
cards = top.to_dict("records")

print("Cards:", len(cards))
print(cards[0])
print(cards[0]["player_display_name"], "plays for", cards[0]["recent_team"])
````

`cards` is a **list of dictionaries**: 10 player cards in a row.
- `cards[0]` is the first card (lists count from 0, Lab 4).
- `cards[0]["recent_team"]` reads the team from that card (key → value, Lab 7).

### 🏈 Your turn: the game!
Read the box before you run it. Almost every line is something you know:

| lines | what they do |
|---|---|
| `secret = random.choice(cards)` | Python pulls one card out of the hat and keeps it secret |
| `answer = secret["player_display_name"]` | the name on the secret card |
| `while guess != answer and tries < max_tries:` | keep asking while the guess is wrong **and** there are tries left |
| `if` / `elif` / `else` | pick the clue: first try → clue 1, second try → clue 2, after that → clue 3 |
| `str(secret[stat])` | **`str( )`** is the twin of `int( )`: it turns a number into text, so `+` can glue it into the clue |
| `answer[0]` | spot 0 of a text is its **first letter**. Counting from 0 works on text too! |
| `guess = input(clue + " Who is it? ")` | show the clue and wait for your guess |

Right now you get only **1** try. Change `max_tries` to **3**, then play. Type the full name with capital letters, like in the table.

````python
import random

stat = "receiving_yards"
max_tries = 1      # how many guesses you get: change it to 3

secret = random.choice(cards)
answer = secret["player_display_name"]
tries = 0
guess = ""

while guess != answer and tries < max_tries:
    if tries == 0:
        clue = "Clue 1: he had " + str(secret[stat]) + " " + stat + "."
    elif tries == 1:
        clue = "Clue 2: he plays for " + secret["recent_team"] + "."
    else:
        clue = "Clue 3: his name starts with " + answer[0] + "."
    guess = input(clue + " Who is it? ")
    tries = tries + 1

if guess == answer:
    print("TOUCHDOWN! It was", answer, "- you got it in", tries, "tries.")
else:
    print("Out of tries. It was", answer)
````

Play it a few times. Python picks a new secret player on every run.
