### A machine with an input
Our cheer is always the same. Let's make it cheer for **any** team. We put an empty box in the round brackets:

````python
def cheer(team):
    print("Go,", team, "!")
    print(team, "fans go wild!")

cheer("Vikings")
cheer("Chiefs")
````

`team` is called a **parameter**: an empty box that gets filled each time you use the machine.

- `cheer("Chiefs")` does it all again, this time with `"Chiefs"` in the box.

It works like the box in a `for` loop (Lab 5): you never fill it with `=` yourself, Python fills it for you.

## Part 2. `return`: the machine hands back an answer 🎁
The `cheer` machine **shows** something. Most machines do a different job: they **work something out and hand the answer back**,
like `len( )` hands back a number. For that we need a new word: **`return`**.

A touchdown is 6 points. **Guess first:** what will this show?

````python
def touchdown_points(touchdowns):
    return touchdowns * 6

points = touchdown_points(3)
print("3 touchdowns:", points, "points")
print("5 touchdowns:", touchdown_points(5), "points")
````

`return` means **"hand this back"**. Take the line `points = touchdown_points(3)` apart, right side first (Lab 2):

| order | what happens |
|---|---|
| 1 | `touchdown_points(3)` puts 3 into the box `touchdowns` |
| 2 | inside the machine, `touchdowns * 6` makes 18 |
| 3 | `return` hands the 18 back to the place where the machine was called |
| 4 | `points =` puts that 18 into a box |

Because the machine **hands back** a number, you can keep it in a box, print it, or do more math with it.

### `print` is not `return`
This machine looks almost the same, but it uses `print` instead of `return`.
**Guess first:** what will the last line show?

````python
def show_points(touchdowns):
    print(touchdowns * 6)

answer = show_points(3)
print("The machine handed back:", answer)
````

The machine showed `18` on the screen, but it handed back... `None`. **`None` is Python's word for "nothing".**

- `print` only **shows** something on the screen. Then it's gone: you can't keep it in a box.
- `return` **hands the answer back**, so the rest of your program can use it.

A machine with no `return` hands back `None`. If you ever see `None` where you expected a number, look for a missing `return`.
This is one of the most common mistakes with machines.

### Two inputs
A machine can have more than one input. Put a comma between them. A field goal is 3 points.
**Guess first:** what are the two scores?

````python
def score(touchdowns, field_goals):
    return touchdowns * 6 + field_goals * 3

print(score(3, 2))
print(score(2, 3))
````

24 and 21. **The order matters:** the first number goes into the first box (`touchdowns`), the second number into the second box (`field_goals`).
`score(3, 2)` is 3 touchdowns and 2 field goals. `score(2, 3)` is 2 touchdowns and 3 field goals.

## Part 3. A fantasy points machine 🏆
In fantasy football, players score points for you with their stats. Our league's rules, for runners and receivers:
- **1 point for every 10 yards** (so yards divided by 10: in Python, `/` means "divided by")
- **6 points for every touchdown**

In 2024 Saquon Barkley ran for **2,005** yards and **13** rushing touchdowns. **Guess first:** about how many points is that?

````python
def fantasy_points(yards, touchdowns):
    return round(yards / 10 + touchdowns * 6, 1)

print(fantasy_points(2005, 13))
````

278.5 points: 200.5 for the yards, plus 78 for the touchdowns.
(`round( ..., 1)` you know from Lab 2: here the `1` means "keep 1 number after the dot".)

Typing stats in by hand is slow. Let's make the machine look them up. This is the season file from Lab 4:
one row is **one whole season** of one player. It loads in a few seconds.

````python
import pandas as pd

seasons = pd.read_csv(
    "https://github.com/nflverse/nflverse-data/releases/download/player_stats/player_stats_season.csv",
    low_memory=False,
)
seasons = seasons[seasons["season_type"] == "REG"]
print("Ready!", len(seasons), "seasons loaded")
````

Now a bigger machine. You give it a **name** and a **year**. It finds that season, adds running and catching together,
and then uses our first machine to turn them into points. **A machine can use another machine!**

````python
def fantasy_season(name, year):
    player = seasons[seasons["player_display_name"] == name]    # only this player (Lab 2)
    row = player[player["season"] == year]                      # only this season
    yards = row["rushing_yards"].sum() + row["receiving_yards"].sum()
    touchdowns = row["rushing_tds"].sum() + row["receiving_tds"].sum()
    return fantasy_points(yards, touchdowns)

print(fantasy_season("Saquon Barkley", 2024))
````

Line by line:

| line | what it does |
|---|---|
| `def fantasy_season(name, year):` | a machine with two boxes: a name and a year |
| `player = ...` and `row = ...` | two steps you know: keep only this player's rows, then only this season |
| `yards = ...` | his rushing yards plus his receiving yards (`.sum()` adds up a column; here the column has just one number in it) |
| `touchdowns = ...` | the same for touchdowns |
| `return fantasy_points(yards, touchdowns)` | hand those two numbers to our first machine, and hand **its** answer back |

It says **318.3**, more than before: now the machine also counts the 278 yards and 2 touchdowns he got **catching** passes.

The boxes `player`, `row`, `yards` and `touchdowns` live **inside** the machine. When the machine finishes, they're thrown away.
Only what comes after `return` gets out.

## Part 4. Rebuild `compare` yourself ⚔️
In Lab 3 you opened up the `compare` machine and fixed its tie bug. Now you'll build it from scratch, your own way, in two small machines.

**Machine 1** adds up one player's whole career (all his seasons since 1999) in one stat:

````python
def career_total(name, stat):
    player = seasons[seasons["player_display_name"] == name]
    return player[stat].sum()

print(career_total("Tom Brady", "passing_tds"))
print(career_total("Peyton Manning", "passing_tds"))
````

**Machine 2** uses machine 1 twice, then decides with `if` / `elif` / `else` (Lab 3).
