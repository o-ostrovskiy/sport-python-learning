## Part 1. Your first player card 🃏
Here is a card for Saquon Barkley, the way it looked when he played for the Giants. Run it:

```python
card = {
    "name": "Saquon Barkley",
    "team": "Giants",
    "number": 26,
    "position": "RB",
}
print(card)
```

That's a **dictionary**. A real dictionary has a word and, next to it, what the word means.
A Python dictionary has a **label** and, next to it, a **value**. The label is called a **key**, because it unlocks the value.

| piece | what it does |
|---|---|
| `{` and `}` | curly brackets: where the dictionary **starts** and **ends** |
| `"team": "Giants"` | one pair: the key `"team"`, a colon, and its value `"Giants"` |
| `:` | the colon here means "goes with": `"team"` goes with `"Giants"` |
| `,` | a comma after each pair, like between the players in a list |

Keys are text, so they're in quotes. Values can be text (`"Giants"`) or numbers (`26`).
We wrote one pair per line so it's easy to read, the same trick as the long list in Lab 4.

### Look something up
In a list you ask for a **spot number**: `my_board[0]`. In a dictionary you ask with the **key**.
**Guess first:** what will these two lines show?

```python
print(card["team"])
print(card["name"], "wears number", card["number"])
```

`card["team"]` means: "from `card`, give me the value that goes with the key `"team"`".
Square brackets after a name always **pick something out of it**: a column from a table (Lab 2), a spot from a list (Lab 4),
and now a value from a dictionary. You never have to remember which spot the team is at. You just ask for it by its label.

What if we ask for a key that isn't on the card? Run this box. It shows an error **on purpose**:

```python
print(card["college"])
```

`KeyError: 'college'`. You met `KeyError` in Lab 2, when a column name wasn't in the table.
It means the same thing here: **"I can't find that key."** There's no `"college"` on this card... yet.

### Add a new fact, change an old one
Both use the same line: the key in square brackets, then `=` and the value.
In 2024 Saquon moved to the Eagles. **Guess first:** what will the card look like after this box?

```python
card["college"] = "Penn State"   # a new key: Python adds it
card["team"] = "Eagles"          # a key that's already there: Python changes its value

print(card)
print("Facts on the card:", len(card))
```

- `"college"` wasn't on the card, so Python **added** the pair at the end.
- `"team"` was already there, so Python **replaced** `"Giants"` with `"Eagles"`. A key can only be on the card **once**.
- `len(card)` counts the pairs: 5 now.

## Part 2. Team codes 🗺️
In our stats tables, teams are written as short codes: `KC`, `MIN`, `PHI`. A dictionary is perfect for turning a code
into a name: the **code is the key**, the **name is the value**. **Guess first:** what will this show?

```python
teams = {"KC": "Chiefs", "MIN": "Vikings", "PHI": "Eagles", "BUF": "Bills", "SF": "49ers"}

print(teams["KC"])
print(teams["SF"])
```

A dictionary is a **lookup table**: you hand it a key, it hands you the value. Short dictionaries fit on one line, like short lists.

### Is the key there?
Asking for a missing key gives a `KeyError`. To check first, use `in` from Lab 4:
with a dictionary, `in` asks **"is this key in it?"**

```python
code = "DAL"

if code in teams:
    print(code, "is the", teams[code])
else:
    print("I don't know", code, "yet")
```

`"DAL"` isn't a key in our dictionary, so `code in teams` is `False` and Python goes to `else`. No red error!

### A loop over a dictionary
A `for` loop (Lab 5) walks through a dictionary **key by key**. Inside the loop, `teams[code]` looks up the value for that key.

```python
for code in teams:
    print(code, "=", teams[code])

print("Teams I know:", len(teams))
```

## Part 3. Cards from the real stats 📊
Typing 32 teams is slow, so here is the whole league, ready-made. Run it.
(The data uses today's codes and names even for old seasons.)

```python
teams = {
    "ARI": "Cardinals", "ATL": "Falcons", "BAL": "Ravens", "BUF": "Bills",
    "CAR": "Panthers", "CHI": "Bears", "CIN": "Bengals", "CLE": "Browns",
    "DAL": "Cowboys", "DEN": "Broncos", "DET": "Lions", "GB": "Packers",
    "HOU": "Texans", "IND": "Colts", "JAX": "Jaguars", "KC": "Chiefs",
    "LA": "Rams", "LAC": "Chargers", "LV": "Raiders", "MIA": "Dolphins",
    "MIN": "Vikings", "NE": "Patriots", "NO": "Saints", "NYG": "Giants",
    "NYJ": "Jets", "PHI": "Eagles", "PIT": "Steelers", "SEA": "Seahawks",
    "SF": "49ers", "TB": "Buccaneers", "TEN": "Titans", "WAS": "Commanders",
}

print("Teams:", len(teams))
print("LAC is the", teams["LAC"])
```

Now the season file from Labs 4 and 6: one row is **one whole season** of one player. It loads in a few seconds.

```python
import pandas as pd

seasons = pd.read_csv(
    "https://github.com/nflverse/nflverse-data/releases/download/player_stats/player_stats_season.csv",
    low_memory=False,
)
seasons = seasons[seasons["season_type"] == "REG"]
print("Ready!", len(seasons), "seasons loaded")
```

Let's find a player's **best season**: keep only his rows, sort them with the most yards first (Lab 2), and keep the top one.

```python
name = "Justin Jefferson"

player = seasons[seasons["player_display_name"] == name]
best = player.sort_values("receiving_yards", ascending=False).head(1)

best[["player_display_name", "recent_team", "position", "season", "receiving_yards"]]
```

`best` is a tiny table with **one row**: his best season. Now we copy the facts from that row onto a card.

```python
code = list(best["recent_team"])[0]

card = {
    "name": name,
    "team": teams[code],
    "position": list(best["position"])[0],
    "best season": int(best["season"].max()),
    "yards": int(best["receiving_yards"].max()),
}
print(card)
print(card["name"], "had his best season in", card["best season"], "with the", card["team"])
```

Each value comes out of the table with tools you already know:

| line | what it does |
|---|---|
| `code = list(best["recent_team"])[0]` | take the team column, turn it into a list (Lab 4), and take spot 0. The list has just one thing in it: `"MIN"`. |
| `"team": teams[code]` | look the code up in our `teams` dictionary: `"MIN"` becomes `"Vikings"`. **A dictionary used inside a dictionary!** |
| `"position": list(best["position"])[0]` | the same trick for the position |
| `int(best["season"].max())` | `.max()` gives the biggest number in the column (there's only one). `int( )` is new: it makes it a **plain whole number**. Without it, Python would show `np.int64(2022)`, a number with a label from the table stuck on it. |
| `"best season"` | a key can have a space in it. It's just text. |

### A card machine
We don't want to write all that again for every player. So let's put it in a machine (Lab 6).
It takes a name and a stat, and **hands back** a finished card.

One word is missing: the 6-letter word that hands the card back.

🔎 **Google it:** `python function return dictionary`

```python
def make_card(name, stat):
    player = seasons[seasons["player_display_name"] == name]
    best = player.sort_values(stat, ascending=False).head(1)
    code = list(best["recent_team"])[0]
    card = {
        "name": name,
        "team": teams[code],
        "position": list(best["position"])[0],
        "best season": int(best["season"].max()),
        "yards": int(best[stat].max()),
    }
    ___ card

print(make_card("Ja'Marr Chase", "receiving_yards"))
print(make_card("Saquon Barkley", "rushing_yards"))
```

Two cards from two short lines! If you see `None` instead of a card, the machine isn't handing anything back: check the missing word.

An `IndexError` here means the machine found **no rows** for that name, so there was no spot 0 to take. Check the spelling.

## Part 4. A deck of cards 🗂️
One card is a dictionary. A **deck** is a list of cards: a list with dictionaries inside.
We start with an empty list `[]`, and a loop adds one card for each name with `.append( )` (Lab 4).

```python
deck = []

for name in ["Justin Jefferson", "Ja'Marr Chase", "CeeDee Lamb"]:
    deck.append(make_card(name, "receiving_yards"))

print("Cards in the deck:", len(deck))
print(deck[0])
```

`deck[0]` is the first card: a whole dictionary sitting at spot 0 of the list.

Now a second loop walks through the deck. Each round, `card` is one dictionary, and we pick out its values by key.
**Guess first:** whose best season had the most yards?

```python
for card in deck:
    print(card["name"], "|", card["team"], "|", card["best season"], "|", card["yards"], "yards")
```
