## Part 1. One line, many players 🔁
In Lab 4 your draft board printed all on one line. To show one player per line, you could write
`print(my_board[0])`, then `print(my_board[1])`, then `print(my_board[2])`... one line of code for every player.
With 50 players that's 50 lines!

A **loop** does it in two lines, however long the board is. **Guess first:** what will this box show?

```python
my_board = ["Justin Jefferson", "Saquon Barkley", "Josh Allen"]

for player in my_board:
    print(player)
```

Read the loop like a sentence: **"for each player in my board, print the player."**

| piece | what it does |
|---|---|
| `for` | starts a loop: "do this for each one" |
| `player` | a box that Python fills for you. Round 1: Justin Jefferson. Round 2: Saquon Barkley. Round 3: Josh Allen. |
| `in my_board` | which list to walk through, from spot 0 to the end |
| `:` | the colon means "here's what to do each round", just like after `if` (Lab 3) |
| the indented line | pushed right with 4 spaces: this is the part that **repeats** |

`player` is a name **we** made up. `for p in my_board:` or `for star in my_board:` would work the same way,
as long as you use the same name on the line below.

**What repeats, and what doesn't?** Only the indented lines belong to the loop.
**Guess first:** how many times will `Draft is over!` show up?

```python
my_board = ["Justin Jefferson", "Saquon Barkley", "Josh Allen"]

for player in my_board:
    print("The pick is in:", player)

print("Draft is over!")
```

Only **once**. The last `print` is not indented, so it's **outside** the loop:
Python runs it one time, after the loop has finished all three rounds.

What if we forget the indent? Run this box. It shows an error **on purpose**:

```python
for player in my_board:
print("The pick is in:", player)
```

The last line says `IndentationError: expected an indented block after 'for' statement on line 1`.
It's the same error as with `if` in Lab 3: Python saw the colon, so it expected a line pushed to the right.
Without the push, it can't tell what to repeat.

## Part 2. `range( )`: Python counts for you 🔢
Sometimes you don't have a list. You just want to do something a number of times, like once for every week.
`range( )` hands the loop numbers, one at a time. **Guess first:** which numbers will this show?

```python
for week in range(5):
    print("Week", week)
```

Five rounds, but it started at **0** and stopped at **4**! Remember Lab 4: Python counts from 0.
`range(5)` means "5 numbers, starting at 0": 0, 1, 2, 3, 4.

There is no "Week 0" in the NFL. So give `range` two numbers: where to **start**, and where to **stop**.
**Guess first:** what is the last week this box shows?

```python
for week in range(1, 6):
    print("Week", week)
```

It stops at **Week 5**, not 6. `range(1, 6)` means: start at 1, and stop **before** 6.

| you write | Python counts |
|---|---|
| `range(5)` | 0, 1, 2, 3, 4 |
| `range(1, 6)` | 1, 2, 3, 4, 5 |
| `range(1, 4)` | 1, 2, 3 |

**The stop number is never included.** So to count up to 5, you stop at 6: one more than the last number you want.

### The draft board, at last
Now we can print the board the way Lab 4 promised: "Pick 1, Pick 2, Pick 3...".

```python
my_board = ["Justin Jefferson", "Saquon Barkley", "Josh Allen", "Ja'Marr Chase", "Bijan Robinson"]

for spot in range(len(my_board)):
    print("Pick", spot + 1, ":", my_board[spot])
```

Take the `for` line apart, from the inside out:

| order | this part | what it does |
|---|---|---|
| 1 | `len(my_board)` | counts the players: 5 |
| 2 | `range(5)` | hands out the numbers 0, 1, 2, 3, 4. Those are exactly the spots on the board! |
| 3 | `for spot in ...` | each round, `spot` holds the next number |

Inside the loop, `my_board[spot]` is the player at that spot (Lab 4).
And `spot + 1` turns Python's counting (0, 1, 2...) into people's counting (1, 2, 3...).

## Part 3. A running total ➕
In 2024, Saquon Barkley had one of the best seasons a running back has ever had.
Here are his rushing yards in his **first 5 games**. Let's add them up the way a scoreboard does: game by game.

**Guess first:** about how many yards after 5 games?

```python
games = [109, 95, 147, 84, 47]

total = 0

for yards in games:
    total = total + yards
    print("This game:", yards, "  Total so far:", total)

print("After 5 games:", total)
```

That's called a **running total**: a box that grows a little every round.

- `total = 0` comes **before** the loop. The box has to exist, and start empty, before we can add to it.
- `total = total + yards` looks strange: how can `total` be on both sides? Remember Lab 2: `=` is not a question,
  it's an order, and Python does the **right side first**.

| order | this part | what it does (round 2, for example) |
|---|---|---|
| 1 | `total + yards` | open the box (109) and add this game's yards (95): that makes 204 |
| 2 | `total =` | put 204 back into the same box. The old 109 is gone. |

- There are **two** indented lines, so both repeat each round. The last `print` is not indented: it runs once, at the end.

### The whole season, from the real data
Typing 5 numbers is fine. Typing a whole season is not. Let's load the games table from Labs 2 and 3
(about 20 seconds):

```python
import pandas as pd

stats = pd.read_csv(
    "https://github.com/nflverse/nflverse-data/releases/download/player_stats/player_stats.csv",
    low_memory=False,
)
stats = stats[stats["season_type"] == "REG"]
print("Ready!", len(stats), "rows loaded")
```

Three small steps, each one you know. Then `list( )` from Lab 4 turns the column into a real list:

```python
name = "Saquon Barkley"

games = stats[stats["player_display_name"] == name]    # only his games (Lab 2)
season = games[games["season"] == 2024]                # only the ones in 2024
yards_list = list(season["rushing_yards"])             # that column, as a real list (Lab 4)

print(yards_list)
print("Games played:", len(yards_list))
```

There are his 16 games, in order. The first five are the numbers from before.

Now the same loop as before walks through the **whole** list. It also counts the games as it goes:
`game = game + 1` is a running total too, it just adds 1 each round.
**Guess first:** did he reach 2,000 yards?

```python
total = 0
game = 0

for yards in yards_list:
    game = game + 1
    total = total + yards
    print("Game", game, ":", yards, "yards.  Season total:", total)

print(name, "finished with", total, "yards")
```

**2,005 yards!** He was only the ninth player in NFL history to run for 2,000 yards in a season.
Look down the "Season total" column: you can see the exact game where he passed 1,000, and the last game where he passed 2,000.

The loop didn't change at all: 5 games or 16, the same three lines do the job. That's the power of a loop.

## Part 4. Count the big games 💯
For a running back, **100 yards** in a game is a big day. How many big days did he have?
We need a loop **and** a decision: an `if` (Lab 3) inside the `for`.

**Guess first:** how many of his 16 games had at least 100 yards?

```python
big_games = 0

for yards in yards_list:
    if yards >= 100:
        big_games = big_games + 1

print("100-yard games:", big_games)
```

Read it like a sentence: **"for each game: if the yards are at least 100, add 1 to the counter."**

- `big_games = 0` starts the counter at zero, before the loop.
- The `if` line is inside the loop, so it's pushed in once (4 spaces). Python asks the question again for every game.
- `big_games = big_games + 1` is inside the loop **and** inside the `if`, so it's pushed in twice (8 spaces).
  It only happens in the rounds where the answer was `True`.

### 🏈 Make it yours: your player's season report
This box has everything from today: a list from the data, a loop, a running total and a counter.
Change the **three** lines at the top to your own player:
- `name`: his name, in quotes, spelled exactly right
- `stat`: `"receiving_yards"`, `"rushing_yards"` or `"passing_yards"`
- `year`: a season he played, from 1999 to 2024

For a quarterback, also change `big = 100` to `big = 300`: for passers, 300 yards is a big day.

```python
name = "Justin Jefferson"   # change to your player
stat = "receiving_yards"    # or "rushing_yards" or "passing_yards"
year = 2022                 # change to a season he played
big = 100                   # a "big game" is this many yards or more

games = stats[stats["player_display_name"] == name]
season = games[games["season"] == year]
yards_list = list(season[stat])

total = 0
big_games = 0

for yards in yards_list:
    total = total + yards
    if yards >= big:
        big_games = big_games + 1

print(name, "in", year)
print("Games:", len(yards_list))
print("Total:", total, "yards")
print("Big games:", big_games)
```

See `Games: 0` and `Total: 0`? Then the list is empty and the loop had nothing to walk through:
check the spelling of the name, and that he really played that season.

Look inside the loop: `total = total + yards` is pushed in once, so it happens **every** round.
The counter line is pushed in twice, so it only happens in the big-game rounds.
