## Part 1. Open a case file 📁
Every investigation starts the same way: get the evidence. This loads the season totals you know from Labs 4, 8 and 9
(one row = one whole season of one player) and opens the drawing toolbox from Lab 9.

```python
import pandas as pd
import matplotlib.pyplot as plt

seasons = pd.read_csv(
    "https://github.com/nflverse/nflverse-data/releases/download/player_stats/player_stats_season.csv",
    low_memory=False,
)
seasons = seasons[seasons["season_type"] == "REG"]
print("Ready!", len(seasons), "player seasons, from", seasons["season"].min(), "to", seasons["season"].max())
```

(`.min()` is the twin of `.max()`: the smallest number in the column, so here the oldest season.)

### My case, as an example
A case file is three things: a **question**, a **player**, and the **stat** that can answer the question.
Here is mine. Read it, then run it.

```python
question = "Was 2024 the best season of Saquon Barkley's career?"
player = "Saquon Barkley"
stat = "rushing_yards"

mine = seasons[seasons["player_display_name"] == player]
print("My question:", question)
print("Seasons found:", len(mine))
mine[["season", "recent_team", "games", stat]]
```

Seven seasons of evidence. `mine[["season", "recent_team", "games", stat]]` shows four columns:
three with their names in quotes, and `stat` **without** quotes, because it's a variable: Python opens the box and finds `"rushing_yards"` inside (Lab 2).

## Part 2. Dig into the numbers ⛏️
Three questions every investigator asks, with three tools from Lab 2. They work on **your** player and **your** stat,
because the box uses your variables `mine` and `stat`.

```python
print("Best season:", mine[stat].max())
print("All seasons added up:", mine[stat].sum())
print("Average season:", round(mine[stat].mean()))
```

- `.max()` is his **best single season**.
- `.sum()` is his **whole career**, added up.
- `.mean()` is the **average**: add up all the seasons and share them out equally. `round( )` drops the digits after the dot.

(For Saquon Barkley's rushing yards these are 2005, 7216 and 1031.)

**When** was the best season? Sort with the biggest on top (Lab 2) and look at the first rows:

```python
best = mine.sort_values(stat, ascending=False)
best[["season", "recent_team", stat]].head(3)
```

### 🏈 Your turn: how many BIG seasons?
What counts as a **big** season depends on the stat. For a runner or a receiver, 1,000 yards is the famous mark.
For a quarterback, maybe 4,000 passing yards or 30 touchdown passes. **You** decide.

The loop is the counter from Lab 5: it walks through the seasons, and every time a number is at or over your bar (`>=`, Lab 3), it adds 1.
Change the `0` to your bar.

```python
bar = 0     # change: how many makes a BIG season for your stat?

big = 0
for number in list(mine[stat]):
    if number >= bar:
        big = big + 1

print(player, "had", big, "seasons with", bar, "or more", stat, "- out of", len(mine))
```

With `bar = 0`, **every** season counts, so the number tells you nothing. A bar only means something if some seasons clear it and some don't.
(For Saquon Barkley and `bar = 1000`: 4 big seasons out of 7.)

## Part 3. Bring in a rival ⚔️
One player's numbers are hard to judge alone. Is 7,216 yards a lot? Compared to **whom**?

Here is a machine like the ones you built in Lab 6. It takes a name and a stat, and gives back (`return`) the career total.
Run the box: the first three lines **build** the machine, and the last line tests it on your player.

```python
def career_total(name, stat):
    rows = seasons[seasons["player_display_name"] == name]
    return rows[stat].sum()

print(player, "-", career_total(player, stat))
```

### 🏈 Your turn: pick the rival
Change `"?"` to a player worth comparing with yours: someone who plays the **same position**, so the same stat fits both.
The `if` / `elif` / `else` from Lab 3 then says who leads. (The `else` is there for a tie, the bug you fixed in Lab 3!)

```python
rival = "?"     # change: the player you compare with

mine_total = career_total(player, stat)
rival_total = career_total(rival, stat)
print(player, "-", mine_total)
print(rival, "-", rival_total)

if mine_total > rival_total:
    print(player, "leads by", mine_total - rival_total)
elif rival_total > mine_total:
    print(rival, "leads by", rival_total - mine_total)
else:
    print("It is a tie!")
```

If your rival shows `0`, Python found no seasons for that name: check the spelling.

### Is it a fair fight?
Careful, investigator. A player with 12 seasons has had many more chances to pile up yards than a player with 4.
Adding up whole careers is only fair when both played about the same number of seasons.

A fairer number is the **average season**: the career total divided by the number of seasons. `/` means "divide", share it out equally.
The loop runs the same lines for both players, so you don't have to write them twice.
The `if` inside is a safety check: nobody can divide by 0 seasons, not even Python.

```python
for name in [player, rival]:
    rows = seasons[seasons["player_display_name"] == name]
    total = career_total(name, stat)
    if len(rows) == 0:
        print(name, "- no seasons found. Check the spelling!")
    else:
        print(name, "-", len(rows), "seasons,", total, "in all,", round(total / len(rows)), "in an average season")
```

Did the leader change? Sometimes the player with the smaller career total has the better average season: he just hasn't played as long.
A good investigator says **which** number the answer is based on.
