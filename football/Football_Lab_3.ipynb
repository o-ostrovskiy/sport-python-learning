## Part 1. Yes or no? 🤔
Python can answer yes-or-no questions. It answers with **`True`** (yes) or **`False`** (no).

**Guess first:** what will these two lines show?

````python
print(10 > 7)
print(3 > 7)
````

- `10 > 7` asks: "is 10 **bigger than** 7?" Yes → `True`.
- `3 > 7` asks: "is 3 bigger than 7?" No → `False`.

`True` and `False` are special Python words: they start with a capital letter and have **no quotes**.

Here are the question symbols:

| symbol | the question it asks |
|---|---|
| `>` | is it bigger than? |
| `<` | is it smaller than? |
| `>=` | is it **at least**? (bigger, or the same) |
| `<=` | is it **at most**? (smaller, or the same) |
| `==` | is it the same? (remember from Lab 2: two `=`) |

A first down needs **at least** 10 yards. **Guess first:** the team gained exactly 10 yards.
Which line says `True`?

````python
yards_to_go = 10
yards_gained = 10

print(yards_gained > yards_to_go)
print(yards_gained >= yards_to_go)
````

`10 > 10` is `False`: 10 is not **bigger** than 10.
`10 >= 10` is `True`: 10 is **at least** 10. That's why referees use "at least", and so will we.

## Part 2. `if`: the first-down referee 🏈
Now Python makes a decision. Read the code like a sentence:
**"if the yards gained are at least the yards to go, say First down!"**

````python
yards_to_go = 10
yards_gained = 12

if yards_gained >= yards_to_go:
    print("First down! 🏈")
````

Three new things:
- **`if`** starts a question. If the answer is `True`, Python does what's underneath.
  If it's `False`, Python skips it.
- The **colon** `:` at the end of the `if` line means "here's what to do".
- The line underneath is pushed to the right with **4 spaces**. That push is called an **indent**,
  and it shows Python which lines belong to the `if`. In Colab, press Enter after the colon
  and it adds the spaces for you.

What if we forget the colon? Run this box. It shows an error **on purpose**:

````python
if yards_gained >= yards_to_go
    print("First down! 🏈")
````

The last line says `SyntaxError: expected ':'`. Python is telling you exactly what's missing: the colon.

### `else`: what to do otherwise
`else` means **"otherwise"**. It runs only when the `if` question was `False`.
Look: `else` has a colon too, and its line is indented too.

````python
yards_to_go = 10
yards_gained = 7

if yards_gained >= yards_to_go:
    print("First down! 🏈")
else:
    print("Not yet. Next down!")
````

## Part 3. `elif`: the 4th-down coach 🧢
It's 4th down. The coach has **three** choices: go for it, kick a field goal, or punt.
Two choices need `if` and `else`. For more, we use **`elif`**, short for "else if": *otherwise, if...*

Our coach bot's rules:
- 1 yard or less to go → **go for it**
- otherwise, 35 yards or less from the end zone → **kick a field goal**
- otherwise → **punt**

````python
yards_to_go = 1
yards_to_end_zone = 60

if yards_to_go <= 1:
    print("Go for it! 💪")
elif yards_to_end_zone <= 35:
    print("Kick a field goal! 🥅")
else:
    print("Punt it away. 🦵")
````

How Python reads it: it asks the questions **from top to bottom**. The **first** `True` wins,
and Python skips everything below it. If nothing is `True`, it does the `else`.

(Real coaches also think about the score and the clock. Our bot keeps it simple.)

## Part 4. Rate your player's best game ⭐
Let's use `if` on real stats. First load the table, the same way as in Lab 2 (about 20 seconds):

````python
import pandas as pd

stats = pd.read_csv(
    "https://github.com/nflverse/nflverse-data/releases/download/player_stats/player_stats.csv",
    low_memory=False,
)
stats = stats[stats["season_type"] == "REG"]
print("Ready!", len(stats), "rows loaded")
````

This box finds a player's best game, like in Lab 2, and then **rates** it:
200 yards or more is a monster game, 100 or more is a big game.

````python
name = "Travis Kelce"

games = stats[stats["player_display_name"] == name]
best = games["receiving_yards"].max()
print(name, "best game:", best, "yards")

if best >= 200:
    print("MONSTER GAME! 🔥")
elif best >= 100:
    print("Big game! 💥")
else:
    print("Solid, but no 100-yard game yet.")
````

## Part 5. Inside the `compare` machine 🔧
Remember `compare` from Lab 2? Now you can read it. Here it is, line by line:

| line | what it means |
|---|---|
| `def compare(player1, player2, stat):` | "teach Python a new command called `compare`; it needs two players and a stat" |
| `total1 = ...` and `total2 = ...` | add up each player's whole career in that stat |
| `if total1 > total2:` | is player 1's total bigger? |
| `else:` | otherwise... |

Run it, and then try the test below it.

````python
def compare(player1, player2, stat):
    total1 = stats[stats["player_display_name"] == player1][stat].sum()
    total2 = stats[stats["player_display_name"] == player2][stat].sum()
    print(player1, ":", total1)
    print(player2, ":", total2)
    if total1 > total2:
        print("Winner:", player1, "🏆")
    else:
        print("Winner:", player2, "🏆")
````

**The test:** compare Tom Brady with... Tom Brady. Their numbers are exactly the same.
What **should** the machine say? Run it and see what it **does** say.

````python
compare("Tom Brady", "Tom Brady", "passing_tds")
````

🐞 **You found a bug!** When the totals are the same, `total1 > total2` is `False`,
so the machine goes to `else` and picks player 2 as the winner. But it's a tie!
