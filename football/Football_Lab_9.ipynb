## Part 1. Your first bar chart 📊
A bar chart needs two lists: the **names** that go along the bottom, and the **numbers** that say how tall each bar is.
These are the top 3 receivers of 2024 and their receiving yards.

**Guess first:** whose bar will be the tallest?

```python
import matplotlib.pyplot as plt

players = ["Ja'Marr Chase", "Justin Jefferson", "Brian Thomas"]
yards = [1708, 1533, 1282]

plt.bar(players, yards)
plt.show()
```

| piece | what it does |
|---|---|
| `import matplotlib.pyplot as plt` | opens Python's drawing toolbox and gives it the short name `plt` (like `import pandas as pd` in Lab 2). You saw it in Lab 1 |
| `plt.bar(players, yards)` | draws one **bar** for each player. First list = names along the bottom, second list = how tall |
| `plt.show()` | "I'm done drawing, show the picture" |

The first name goes with the first number, the second with the second, and so on. That's why the **order** in both lists matters.

### A chart needs words
Look at the chart again. What do the numbers on the left side count? Yards? Touchdowns? Dollars?
**Nobody can tell.** A chart without words is a riddle.

### 🏈 Your turn
Give the chart a **title** (what is this chart about?) and a **label** for the left side (what do the numbers count?).
Write your words between the empty quotes `""`.

```python
players = ["Ja'Marr Chase", "Justin Jefferson", "Brian Thomas"]
yards = [1708, 1533, 1282]

plt.bar(players, yards)
plt.title("")      # what is this chart about? write it between the quotes
plt.ylabel("")     # what do the numbers on the left count?
plt.show()
```

- `plt.title("...")` writes words **on top** of the chart.
- `plt.ylabel("...")` writes words along the **left side**. The left side is called the **y** side, the bottom is the **x** side
  (so there's a `plt.xlabel` too, you'll use it in Part 3).

Now anyone can read your chart without asking you.

## Part 2. From a dictionary to a chart 🃏
In Lab 7 you kept stats in a **dictionary**: name → number. Here are the touchdown passes of three quarterbacks in 2024.
A chart wants two lists, so a `for` loop (Lab 5) walks through the dictionary and fills them with `.append( )` (Lab 4).

**Guess first:** what will the two lists look like?

```python
tds = {
    "Joe Burrow": 43,
    "Lamar Jackson": 41,
    "Baker Mayfield": 41,
}

names = []
numbers = []
for name in tds:
    names.append(name)
    numbers.append(tds[name])

print(names)
print(numbers)
```

| line | what it does |
|---|---|
| `names = []` and `numbers = []` | two **empty lists**, ready to be filled |
| `for name in tds:` | walk through the dictionary, one **key** (one name) at a time |
| `names.append(name)` | add the name to the first list |
| `numbers.append(tds[name])` | look up his number in the dictionary, add it to the second list |

Two lists, same order, same length. Just what `plt.bar` wants:

```python
plt.bar(names, numbers, color="orange")
plt.title("Touchdown passes in 2024")
plt.ylabel("Touchdown passes")
plt.show()
```

`color="orange"` is a **setting**, like `ascending=False` in Lab 2. Try `"green"`, `"red"` or `"purple"` later.

### 🏈 Your turn
Jared Goff threw **37** touchdown passes in 2024. Add him to the dictionary as a new line, right under Baker Mayfield:
`"Jared Goff": 37,`
Then run the box. The loop and the chart don't need to change at all: that's the nice thing about a loop.

```python
tds = {
    "Joe Burrow": 43,
    "Lamar Jackson": 41,
    "Baker Mayfield": 41,
}

names = []
numbers = []
for name in tds:
    names.append(name)
    numbers.append(tds[name])

plt.bar(names, numbers, color="orange")
plt.title("Touchdown passes in 2024")
plt.ylabel("Touchdown passes")
plt.show()
```

Four bars now, and Goff's is the shortest. In `plt.bar(names, numbers)` the **second** list decides how tall the bars are.

## Part 3. A line chart: season by season 📈
A bar chart answers "**who has more?**". A **line chart** answers a different question: "**how did it change over time?**"

Let's load the season totals from Lab 4 (one row = one whole season of one player) and pull out Patrick Mahomes.

```python
import pandas as pd

seasons = pd.read_csv(
    "https://github.com/nflverse/nflverse-data/releases/download/player_stats/player_stats_season.csv",
    low_memory=False,
)
seasons = seasons[seasons["season_type"] == "REG"]

mahomes = seasons[seasons["player_display_name"] == "Patrick Mahomes"]
mahomes[["season", "recent_team", "passing_yards"]]
```

Eight seasons, eight numbers. (In 2017 he was the backup and played one game, so that number is tiny.)
**Guess first:** which season was his best? Now let's see it as a picture.

```python
plt.plot(mahomes["season"], mahomes["passing_yards"], marker="o")
plt.title("Patrick Mahomes: passing yards, season by season")
plt.xlabel("Season")
plt.ylabel("Passing yards")
plt.show()
```

| piece | what it does |
|---|---|
| `plt.plot(...)` | draws a **line** instead of bars. First the column for the bottom (seasons), then the column for the height (yards) |
| `marker="o"` | a setting: put a dot on every season, so you can see each one |
| `plt.xlabel("Season")` | words along the **bottom** (the x side) |

Read it from left to right: older seasons on the left, newer on the right. When the line goes **up**, he threw for more yards than the year before.
The highest dot is **2022**: 5,250 yards.

### 🏈 Your turn: your team, season by season
One team has many players, so we add them up. This box uses the **running-total idea** from Lab 5:
for every year, keep that year's rows and add up the passing yards with `.sum()` (Lab 2).

It's set to the Chiefs (`"KC"`). **Change `"KC"` to your team's short name** and run it.

`ARI ATL BAL BUF CAR CHI CIN CLE DAL DEN DET GB HOU IND JAX KC LA LAC LV MIA MIN NE NO NYG NYJ PHI PIT SEA SF TB TEN WAS`
(`LA` is the Rams, `LAC` the Chargers, `LV` the Raiders.)

```python
team = "KC"      # change to your team's short name

rows = seasons[seasons["recent_team"] == team]

years = []
totals = []
for year in range(2015, 2025):
    one_year = rows[rows["season"] == year]
    years.append(year)
    totals.append(one_year["passing_yards"].sum())

plt.bar(years, totals)
plt.title(team + ": passing yards, season by season")
plt.xlabel("Season")
plt.ylabel("Passing yards by the whole team")
plt.show()
```

- `range(2015, 2025)` gives the years 2015, 2016, ... up to **2024**. Remember from Lab 5: `range` stops one **before** the second number.
- `team + ": passing yards..."` glues your team's name into the title, so the title is always right.
- Can you spot the year a new quarterback arrived?

(One small thing: a player who changed teams in the middle of a season is counted with his **last** team that year.
So a number can be a little off, but the picture is right.)

## Part 4. Two careers, side by side ⚔️
The best thing about a line chart: you can draw **two lines** on the same picture and compare them.
Two `plt.plot` lines = two lines on the chart.

```python
one = seasons[seasons["player_display_name"] == "Patrick Mahomes"]
two = seasons[seasons["player_display_name"] == "Josh Allen"]

plt.plot(one["season"], one["passing_yards"], marker="o")
plt.plot(two["season"], two["passing_yards"], marker="o")
plt.title("Mahomes and Allen: passing yards")
plt.xlabel("Season")
plt.ylabel("Passing yards")
plt.show()
```

Two lines, two colors. But **which line is Mahomes?** The chart doesn't say. It's a riddle again!

### 🏈 Your turn: name the lines
Two steps fix it:
1. Each `plt.plot` gets a setting `label="..."` with the player's name. That's already done for you below.
2. One more command draws the little box that says which color is who. One word is missing.
   It's 6 letters, and it's the same word a map uses for the box that explains its colors and symbols.

🔎 **Google it:** `matplotlib show label box`

```python
one = seasons[seasons["player_display_name"] == "Patrick Mahomes"]
two = seasons[seasons["player_display_name"] == "Josh Allen"]

plt.plot(one["season"], one["passing_yards"], marker="o", label="Patrick Mahomes")
plt.plot(two["season"], two["passing_yards"], marker="o", label="Josh Allen")
plt.___()
plt.title("Mahomes and Allen: passing yards")
plt.xlabel("Season")
plt.ylabel("Passing yards")
plt.show()
```

Now the chart tells the whole story: Mahomes started a year earlier, and his line is on top almost every season.
Look closely at 2023: that year Allen's dot is a tiny bit higher.

### 🏈 Make it yours: your own matchup
Pick **your** two players and the stat to compare. Change the three lines at the top, and write the left-side label in plain words.

Stats you can use: `"passing_yards"`, `"passing_tds"`, `"rushing_yards"`, `"rushing_tds"`, `"receiving_yards"`, `"receiving_tds"`, `"receptions"`
(tds = touchdowns, receptions = catches). Pick a stat that fits your players: a receiver has almost no passing yards!

If a line is missing, check the spelling of the name: capital letters matter. The data goes from 1999 to 2024.

One new line in this box: `plt.xticks(...)` puts a mark on the bottom for **every season** your players played.
Inside it, `list(one["season"]) + list(two["season"])` glues the two lists of seasons into one long list
(`+` glues lists, just like it glues text). `rotation=90` turns the years sideways, so they never bump into each other.
Without this line Python sometimes invents marks like `2022.5`, and there is no season 2022.5!

```python
player_1 = "Patrick Mahomes"    # change to your first player
player_2 = "Josh Allen"         # change to your second player
stat = "passing_yards"          # change the stat if you like
words = "Passing yards"         # the label for the left side, in plain words

one = seasons[seasons["player_display_name"] == player_1]
two = seasons[seasons["player_display_name"] == player_2]

plt.plot(one["season"], one[stat], marker="o", label=player_1)
plt.plot(two["season"], two[stat], marker="o", label=player_2)
plt.xticks(list(one["season"]) + list(two["season"]), rotation=90)
plt.legend()
plt.title(player_1 + " and " + player_2)
plt.xlabel("Season")
plt.ylabel(words)
plt.show()
```
