They are opposites. `int` is short for **integer**, the math word for a whole number. `str` is short for **string**,
Python's word for text (letters on a string, like beads).

## Part 2. `while`: keep going ⚾
In Lab 5 you used a `for` loop: "do this once **for each** thing in the list". You knew how many times it would go around.

In a guessing game nobody knows how many tries it will take. For that, Python has a second kind of loop: **`while`**.
It means **"as long as this is true, keep going"**.

**Guess first:** how many lines will this box print?

````python
strikes = 0

while strikes < 3:
    strikes = strikes + 1
    print("Strike", strikes)

print("You're out! ⚾")
````

Four lines: three strikes, then "You're out!". Here is what Python did:

| piece | what it does |
|---|---|
| `while strikes < 3:` | a question, like after `if`: "is `strikes` less than 3?" If `True`, do the indented lines. Ends with a colon `:` |
| the indented lines | the part that repeats (just like under `for` and `if`) |
| `strikes = strikes + 1` | one more strike. You know this from Lab 5: take what's in the box, add 1, put it back |
| back to the top | after the indented lines Python **jumps back up** and asks the question again |
| `print("You're out! ⚾")` | not indented, so it runs **once**, after the loop is over |

Round by round: strikes is 0 → "less than 3?" yes → Strike 1. Then 1 → yes → Strike 2. Then 2 → yes → Strike 3.
Then 3 → "less than 3?" **no** → the loop stops.

**The difference from `if`:** `if` asks its question **once**. `while` asks it **again and again**.

### ⚠️ The loop that never stops
What if you forget the line `strikes = strikes + 1`?

```
strikes = 0
while strikes < 3:
    print("Strike", strikes)
```

Then `strikes` stays 0 forever, the answer to "less than 3?" is always yes, and Python prints `Strike 0`
thousands of times and **never stops**. Nothing breaks, but the box runs forever.
If that ever happens to you, press the **■ stop button** where the ▶ was.
So remember: **something inside a `while` loop must change, or it never ends.**

### Two small helpers for the game
Our game loop will need to ask **two** questions at once: "is the guess still wrong?" **and** "are there tries left?"
**Guess first:** five answers, `True` or `False`?

````python
print("Babe Ruth" == "Babe Ruth")
print("Babe Ruth" != "Babe Ruth")
print("Babe Ruth" != "Hank Aaron")

print(2 < 3 and 5 < 9)
print(2 < 3 and 5 < 1)
````

`True`, `False`, `True`, then `True`, `False`.

| piece | what it asks |
|---|---|
| `!=` | "are these **not** the same?" It's the opposite of `==` (Lab 3). Read it as "is not" |
| `and` | sits between two questions and says `True` only when **both** are `True` |

The last line is `False` because `5 < 1` isn't true, and `and` needs both sides.

### Now the quiz, with 3 tries

````python
tries = 0
guess = 0

___ guess != 60 and tries < 3:
    guess = int(input("How many home runs did Babe Ruth hit in 1927? "))
    tries = tries + 1
    if guess < 60:
        print("More than that!")
    elif guess > 60:
        print("Fewer than that!")

if guess == 60:
    print("Yes! 60 home runs 🏆 Tries used:", tries)
else:
    print("Out of tries. It was 60.")
````

Read the `while` line slowly: keep going while `guess != 60` (still wrong) `and` `tries < 3` (tries left).
So this loop can end in two ways:
- you guess right → `guess != 60` becomes `False`
- you run out of tries → `tries < 3` becomes `False`

Either way the loop is over, and the `if` at the bottom (not indented, so it runs once) says how it went.

Why `guess = 0` at the top? The `while` line looks into the box `guess` **before** you have typed anything,

## Part 3. `random`: Python picks 🎲
A guessing game is no fun if you already know the answer. Python has to pick **by surprise**.
Run this box **five times** and watch.

````python
import random

hitters = ["Shohei Ohtani", "Aaron Judge", "Juan Soto", "Cal Raleigh", "Bobby Witt Jr."]

print(random.choice(hitters))
````

A different name comes out each time (sometimes the same one twice in a row: that's luck).

| piece | what it does |
|---|---|
| `import random` | opens Python's toolbox for luck and chance, like `import pandas` opens the toolbox for tables (Lab 2) |
| `random.choice(hitters)` | picks **one** thing from the list by chance, like pulling a name out of a hat |
| the dot | "from the `random` toolbox, use `choice`" |

**Random** means nobody can know ahead of time what comes out. Not even you, the programmer!

Now let's keep the pick a **secret**: put it into a variable and **don't print it**.
Run the box and try to guess which of the 5 hitters Python picked. Spell the name exactly as in the list.

````python
hitters = ["Shohei Ohtani", "Aaron Judge", "Juan Soto", "Cal Raleigh", "Bobby Witt Jr."]
secret = random.choice(hitters)

guess = input("I picked one of the 5 hitters. Who is it? ")

if guess == secret:
    print("YES! 🎉 It was", secret)
else:
    print("No, it was", secret)
````

With 5 players you'll be right about 1 time in 5. That's pure luck. A good game gives **clues**.
And for clues we need real stats.

## Part 4. The real game 🕵️
Load the data with the hidden box, the same way as always (click ▶, about 20 seconds).
Remember: one row is one player's whole season.

````python
#@title ⚾ Load the MLB data, 1871 to today (click ▶ and wait about 20 seconds)
import datetime
import unicodedata

import pandas as pd
import requests

LAHMAN = "https://raw.githubusercontent.com/daviddalpiaz/pylahman/main/data-raw/"
MLB = "https://statsapi.mlb.com/api/v1/"
STATS = ["G", "AB", "H", "HR", "RBI", "SB", "BB", "SO"]


def plain(text):
    # "Acuña" -> "Acuna": kids type names on an English keyboard
    return unicodedata.normalize("NFKD", str(text)).encode("ascii", "ignore").decode()


def short(text):
    # "Acuña Jr." -> "acu": the start of a last name, to match players across the two sources
    return plain(text).lower()[:3]


# History: the Lahman Baseball Database (SABR, CC BY-SA 3.0). One row per player, season and team.
batting = pd.read_csv(LAHMAN + "Batting.csv")
people = pd.read_csv(LAHMAN + "People.csv")
teams = pd.read_csv(LAHMAN + "Teams.csv")[["yearID", "teamID", "name"]].rename(columns={"name": "team"})
batting = batting.merge(teams, on=["yearID", "teamID"], how="left")

# A player traded during a season has a row for each team: add them up and keep his last team.
per_season = batting.groupby(["playerID", "yearID"], sort=False)
history = per_season[STATS].sum(min_count=1).join(per_season["team"].last()).reset_index()
people["name"] = people["nameFirst"] + " " + people["nameLast"]
history = history.merge(people[["playerID", "name"]], on="playerID").rename(columns={"yearID": "season"})

# The seasons the history book doesn't have yet come live from MLB (statsapi.mlb.com).
lahman_ids = {
    (int(p.birthYear), int(p.birthMonth), int(p.birthDay), short(p.nameLast)): p.playerID
    for p in people.dropna(subset=["birthYear", "birthMonth", "birthDay"]).itertuples()
}
mlb_names, live = {}, []
try:
    for season in range(int(history["season"].max()) + 1, datetime.date.today().year + 1):
        url = MLB + f"stats?stats=season&group=hitting&season={season}&sportId=1&playerPool=ALL&limit=5000"
        found = requests.get(url, timeout=60).json().get("stats", [])
        hitters = found[0]["splits"] if found else []
        if not hitters:
            continue
        roster = requests.get(MLB + f"sports/1/players?season={season}", timeout=60).json()["people"]
        born = {p["id"]: p.get("birthDate", "") for p in roster}
        for h in hitters:
            p, s = h["player"], h["stat"]
            y, m, d = (born.get(p["id"]) or "0-0-0").split("-")
            player_id = lahman_ids.get((int(y), int(m), int(d), short(p["lastName"])), f"mlb{p['id']}")
            mlb_names[player_id] = p["fullName"]
            live.append({
                "playerID": player_id, "season": season, "team": h.get("team", {}).get("name"), "name": p["fullName"],
                "G": s["gamesPlayed"], "AB": s["atBats"], "H": s["hits"], "HR": s["homeRuns"], "RBI": s["rbi"],
                "SB": s["stolenBases"], "BB": s["baseOnBalls"], "SO": s["strikeOuts"],
            })
except (requests.RequestException, ValueError, KeyError) as problem:
    live = []
    print("MLB's live stats didn't answer, so the newest season is missing today. Everything else works.", problem)

stats = pd.concat([history, pd.DataFrame(live)], ignore_index=True)
# Players active today get the name MLB uses now on their whole career ("Vladimir Guerrero Jr."), without accents.
stats["name"] = stats["playerID"].map(mlb_names).fillna(stats["name"]).map(plain)
# Whole numbers (114, not 114.0); a stat nobody counted back then stays empty: <NA>.
stats[STATS] = stats[STATS].astype("Int64")
stats = stats.sort_values(["playerID", "season"], ignore_index=True)
stats = stats[["name", "season", "team"] + STATS + ["playerID"]]
print("Ready!", len(stats), "rows, seasons", stats["season"].min(), "to", stats["season"].max())
````

First, the players Python may pick from: the **top 10 home run hitters** of the newest season, as a list.
You wrote these exact lines in Lab 4.

````python
newest = stats["season"].max()
last = stats[stats["season"] == newest]
best = last.sort_values("HR", ascending=False)
players = list(best["name"].head(10))

print(newest)
print(players)
````

Next, a small machine that looks up one fact about one player. You know how to build machines from Lab 6
(`def`, inputs, `return`):

````python
def clue(name, column):
    row = last[last["name"] == name]
    return row[column].max()


print(players[0])
print(clue(players[0], "team"))
print(clue(players[0], "HR"))
````

| line | what it does |
|---|---|
| `def clue(name, column):` | a machine called `clue` with two inputs: a player's name and a column of the table |
| `row = last[last["name"] == name]` | keep only that player's row from the newest season (Lab 2) |
| `return row[column].max()` | hand back what's in that column. There is only one row, so `.max()` just gives us that one value |

The three `print` lines try the machine out on `players[0]`, the first player on the list: his name, his team, his home runs.

Now the game itself. Read it slowly before you run it. Every piece is something you know:

````python
secret = random.choice(players)

clues = [
    "Clue 1: he plays for the " + clue(secret, "team") + ".",
    "Clue 2: he hit " + str(clue(secret, "HR")) + " home runs in " + str(newest) + ".",
    "Clue 3: his name starts with " + secret[0] + ".",
]

tries = 0
guess = ""

while guess != secret and tries < 3:
    guess = input(clues[tries] + " Who is it? ")
    tries = tries + 1

if guess == secret:
    print("⚾ YES! You got it! Tries used:", tries)
else:
    print("Out of tries. The mystery player was", secret)
````

How it works, top to bottom:

| piece | what it does |
|---|---|
| `secret = random.choice(players)` | Python secretly picks one of the 10 players (Part 3) |
| `clues = [ ... ]` | a **list** of 3 clues, written over several lines (Lab 4) |
| `"..." + clue(secret, "team") + "."` | glues text together with `+`. The team is already text |
| `str(clue(secret, "HR"))` | home runs are a **number**, so `str( )` turns them into text before gluing (Part 1) |
| `secret[0]` | spot 0 of a name is its **first letter**. Text works like a list of letters! (Lab 4) |
| `guess = ""` | an empty text, so the box `guess` exists before the first question |
| `while guess != secret and tries < 3:` | keep asking while the guess is wrong **and** tries are left (Part 2) |
| `clues[tries]` | try 0 shows clue 0, try 1 shows clue 1, try 2 shows clue 2: a new clue each time (Lab 4: counting from 0) |
| the `if` at the end | not indented, so it runs once, after the loop: did you get it? |

Type the full name with capital letters, exactly as in the list above: `Pete Alonso`, not `pete alonso`.
To Python those are different (Lab 2).

Run the box again: a new mystery player every time!

- `how_many`: how many top players Python picks from. More players = harder

````python
stat = "HR"      # change to "SB", "H" or "BB"
how_many = 10    # change to make it harder or easier
chances = 3      # how many tries

best = last.sort_values(stat, ascending=False)
players = list(best["name"].head(how_many))
secret = random.choice(players)

clues = [
    "Clue 1: he plays for the " + clue(secret, "team") + ".",
    "Clue 2: his " + stat + " in " + str(newest) + ": " + str(clue(secret, stat)) + ".",
    "Clue 3: his name starts with " + secret[0] + ".",
]

tries = 0
guess = ""

while guess != secret and tries < chances:
    guess = input(clues[tries] + " Who is it? ")
    tries = tries + 1

if guess == secret:
    print("⚾ YES! You got it! Tries used:", tries)
else:
    print("Out of tries. The mystery player was", secret)
````
