## Part 1. One batter at a time 🔁
In Lab 4, `print(lineup)` showed the whole lineup squeezed onto one line. A real lineup card has one batter per line.
You could write `print(lineup[0])`, `print(lineup[1])`... nine times. Or you can write **two lines** and let Python do the repeating.

**Guess first:** what will this box show?

```python
lineup = [
    "Shohei Ohtani",
    "Aaron Judge",
    "Juan Soto",
    "Cal Raleigh",
    "Bobby Witt Jr.",
    "Jose Ramirez",
    "Kyle Schwarber",
    "Julio Rodriguez",
    "Pete Alonso",
]

for batter in lineup:
    print(batter)
```

Nine batters, one per line, and you only wrote `print` **once**. That's a **loop**. Here's how to read it:

| piece | what it does |
|---|---|
| `for` | "do this **for** each one" |
| `batter` | a box (a variable, Lab 2). The loop puts the **next** player into it each time around. You pick its name. |
| `in lineup` | which list to walk through, from the top to the bottom |
| `:` | the colon, like after `if` in Lab 3: "here comes what to do" |
| the pushed-in line | what to do **each time**. Pushed in 4 spaces, just like under `if`. |

So Python does this: put `"Shohei Ohtani"` into `batter`, run the pushed-in line. Put `"Aaron Judge"` into `batter`,
run the pushed-in line. And so on, until the list runs out.

**What's inside the loop, and what's not?** Only the **pushed-in** lines repeat.
**Guess first:** how many times will this box say `Play ball!`?

```python
for batter in lineup:
    print("Now batting:", batter)
print("Play ball!")
```

Only **once**, at the very end. `print("Play ball!")` is **not** pushed in, so it's not part of the loop.
Python finishes all 9 trips around the loop first, then moves on.

What if you forget to push the line in? Run this box. It shows an error **on purpose**:

```python
for batter in lineup:
print(batter)
```

The last line says `IndentationError: expected an indented block after 'for' statement`.
You saw this error with `if` in Lab 3. It means: "after the `for` line with its colon, I need at least one pushed-in line.
Otherwise I don't know **what** to repeat."

## Part 2. Counting innings with `range( )` 🔢
Sometimes you don't have a list. You just want to count: inning 1, inning 2, inning 3...
Python has a counting machine for that: **`range( )`**.

**Guess first:** which numbers will this box show? (Remember how Python counts, from Lab 4.)

```python
for inning in range(9):
    print("Inning", inning)
```

`range(9)` gives **9 numbers**, but it starts at **0**, like the spots in a list: 0, 1, 2 ... 8.
There's no "inning 0" in baseball! Give `range` **two** numbers and it starts where you say:

```python
for inning in range(1, 10):
    print("Inning", inning)
```

`range(1, 10)` means: "start at 1, and stop **before** 10". So the last number is 9.
The second number is always one **more** than where you want to end. (Strange, but you get used to it.)

Now the lineup card from the end of Lab 4, with batting order numbers.
**Guess first:** why does it say `spot + 1`?

```python
for spot in range(9):
    print(spot + 1, lineup[spot])
```

- `range(9)` counts 0 to 8: exactly the spot numbers of a 9-player list. So `lineup[spot]` is each batter in turn.
- A lineup card says 1 to 9, not 0 to 8. So we show `spot + 1`. The math from Lab 1 still works inside a loop!

## Part 3. Keeping score: a running total ➕
The scoreboard doesn't only show each inning. It also keeps the **score so far**. A loop can do that too.

**Guess first:** what's the last number this box will show?

```python
dodgers = [0, 0, 0, 0, 5, 0, 0, 2, 0]

total = 0
for runs in dodgers:
    total = total + runs
    print("Score so far:", total)

print("Final score:", total)
```

The Dodgers scored **7**. Here's the trick, line by line:

| line | what it does |
|---|---|
| `total = 0` | **before** the loop: an empty scoreboard. A box called `total` with 0 in it. |
| `for runs in dodgers:` | walk through the innings. `runs` holds the runs of one inning. |
| `total = total + runs` | add this inning's runs to the score |
| `print("Final score:", total)` | not pushed in, so it runs **once**, after the last inning |

**`total = total + runs` looks strange.** How can `total` be the same as `total` plus something?
Remember Lab 2: one `=` doesn't mean "the same". It means **"put into the box"**.
Python always does the **right side first**: take what's in `total` now, add `runs`. Then it puts the answer **back** into `total`.
The old number is gone, the new score is in. That's exactly what a scoreboard does.

This is called a **running total**: a total that grows as you go.

### A whole career, season by season
A game has 9 innings. A career has many seasons. Same loop! Load the data with the hidden box,
the same way as in Lab 4 (click ▶, about 20 seconds). Remember: one row is one player's whole season.

```python
#@title ⚾ Load the MLB data, 1871 to today (click ▶ and wait about 20 seconds)
import datetime
import unicodedata

import pandas as pd
import requests

LAHMAN = "https://raw.githubusercontent.com/daviddalpiaz/pylahman/main/data-raw/"
MLB = "https://statsapi.mlb.com/api/v1/"
STATS = ["G", "AB", "H", "HR", "RBI", "SB", "BB", "SO"]


def plain(text):
    # "Acuña" -> "Acuna": kids type names on an English keyboard
    return unicodedata.normalize("NFKD", str(text)).encode("ascii", "ignore").decode()


def short(text):
    # "Acuña Jr." -> "acu": the start of a last name, to match players across the two sources
    return plain(text).lower()[:3]


# History: the Lahman Baseball Database (SABR, CC BY-SA 3.0). One row per player, season and team.
batting = pd.read_csv(LAHMAN + "Batting.csv")
people = pd.read_csv(LAHMAN + "People.csv")
teams = pd.read_csv(LAHMAN + "Teams.csv")[["yearID", "teamID", "name"]].rename(columns={"name": "team"})
batting = batting.merge(teams, on=["yearID", "teamID"], how="left")

# A player traded during a season has a row for each team: add them up and keep his last team.
per_season = batting.groupby(["playerID", "yearID"], sort=False)
history = per_season[STATS].sum(min_count=1).join(per_season["team"].last()).reset_index()
people["name"] = people["nameFirst"] + " " + people["nameLast"]
history = history.merge(people[["playerID", "name"]], on="playerID").rename(columns={"yearID": "season"})

# The seasons the history book doesn't have yet come live from MLB (statsapi.mlb.com).
lahman_ids = {
    (int(p.birthYear), int(p.birthMonth), int(p.birthDay), short(p.nameLast)): p.playerID
    for p in people.dropna(subset=["birthYear", "birthMonth", "birthDay"]).itertuples()
}
mlb_names, live = {}, []
try:
    for season in range(int(history["season"].max()) + 1, datetime.date.today().year + 1):
        url = MLB + f"stats?stats=season&group=hitting&season={season}&sportId=1&playerPool=ALL&limit=5000"
        found = requests.get(url, timeout=60).json().get("stats", [])
        hitters = found[0]["splits"] if found else []
        if not hitters:
            continue
        roster = requests.get(MLB + f"sports/1/players?season={season}", timeout=60).json()["people"]
        born = {p["id"]: p.get("birthDate", "") for p in roster}
        for h in hitters:
            p, s = h["player"], h["stat"]
            y, m, d = (born.get(p["id"]) or "0-0-0").split("-")
            player_id = lahman_ids.get((int(y), int(m), int(d), short(p["lastName"])), f"mlb{p['id']}")
            mlb_names[player_id] = p["fullName"]
            live.append({
                "playerID": player_id, "season": season, "team": h.get("team", {}).get("name"), "name": p["fullName"],
                "G": s["gamesPlayed"], "AB": s["atBats"], "H": s["hits"], "HR": s["homeRuns"], "RBI": s["rbi"],
                "SB": s["stolenBases"], "BB": s["baseOnBalls"], "SO": s["strikeOuts"],
            })
except (requests.RequestException, ValueError, KeyError) as problem:
    live = []
    print("MLB's live stats didn't answer, so the newest season is missing today. Everything else works.", problem)

stats = pd.concat([history, pd.DataFrame(live)], ignore_index=True)
# Players active today get the name MLB uses now on their whole career ("Vladimir Guerrero Jr."), without accents.
stats["name"] = stats["playerID"].map(mlb_names).fillna(stats["name"]).map(plain)
# Whole numbers (114, not 114.0); a stat nobody counted back then stays empty: <NA>.
stats[STATS] = stats[STATS].astype("Int64")
stats = stats.sort_values(["playerID", "season"], ignore_index=True)
stats = stats[["name", "season", "team"] + STATS + ["playerID"]]
print("Ready!", len(stats), "rows, seasons", stats["season"].min(), "to", stats["season"].max())
```

First, keep only Aaron Judge's rows (your trick from Lab 2), and look at his home runs:

```python
judge = stats[stats["name"] == "Aaron Judge"]
judge[["season", "HR"]]
```

A column of the table works like a list: a loop can walk **down** it, one season at a time.
**Guess first:** about how many home runs has Judge hit in his whole career?

```python
total = 0
for hr in judge["HR"]:
    total = total + hr
    print(hr, "home runs. Career so far:", total)

print("Career home runs:", total)
```

It's the scoreboard loop again. Only the names changed: `runs` became `hr`, and the list became the column `judge["HR"]`.
Through the 2026 season, Judge has hit **386** home runs.

In Lab 2 you added up a column with `.sum()`. Check that it gives the same answer:

```python
print(judge["HR"].sum())
```

The same number! Now you know what `.sum()` does inside: a loop with a running total, just like yours.

## Part 4. Count the big seasons 💪
Hitting **30 home runs** in one season is a big deal. How many times has Judge done it?
We don't want to add up home runs now. We want to **count seasons**: add **1** each time a season is big enough.
For "big enough" you need `if` from Lab 3.

**Guess first:** look at the table above. How many seasons have 30 or more?

```python
count = 0
for hr in judge["HR"]:
    if hr >= 30:
        count = count + 1

print("Seasons with 30 or more home runs:", count)
```

**6** seasons. Read it like this:

| line | what it does |
|---|---|
| `count = 0` | start counting from zero |
| `for hr in judge["HR"]:` | look at each season's home runs |
| `if hr >= 30:` | is this season 30 **or more**? (`>=` from Lab 3) |
| `count = count + 1` | yes → one more big season |

**Why is `count = count + 1` pushed in twice?** The `if` is inside the loop: pushed in once (4 spaces).
The `count` line is inside the `if`, which is inside the loop: pushed in **twice** (8 spaces).
You saw this in Lab 3, inside the `compare` machine.

### 🏆 That's the ballgame!
Today you learned:
- a **`for` loop** repeats the pushed-in lines once for each thing in a list: `for batter in lineup:`
- the box after `for` (`batter`, `runs`, `hr`) gets the **next** thing each time around
- only the **pushed-in** lines repeat. No pushed-in line gives an `IndentationError`
- **`range(9)`** counts 0 to 8, and **`range(1, 10)`** counts 1 to 9
- a **running total**: start with `total = 0`, then `total = total + runs` inside the loop
- to **count**, add 1 inside an `if`: `count = count + 1`
