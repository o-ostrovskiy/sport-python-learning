## Part 1. Your first player card 🃏
**Guess first:** what will this box show?

```python
card = {
    "name": "Aaron Judge",
    "team": "New York Yankees",
    "number": 99,
    "position": "right field",
}

print(card)
```

That's a **dictionary**. A list (Lab 4) keeps things **in order**, and you ask for them by spot number.
A dictionary keeps things **by label**, and you ask for them by label.

| piece | what it does |
|---|---|
| `{` and `}` | curly brackets: where the dictionary **starts** and **ends** |
| `"team"` | a label. Python calls it a **key**. |
| `:` | the colon joins a key to its fact |
| `"New York Yankees"` | the fact. Python calls it a **value**. |
| `,` | a comma after each pair, like in a list |

So a dictionary is made of **pairs**: key `:` value. Text goes in quotes, numbers don't, as always.
We wrote one pair per line, like the long lineup in Lab 4, so it's easy to read.

(It's called a dictionary because it works like a real one: you look up a **word** and get what it means.)

To get one fact, put its key in square brackets. **Guess first:** what will these two lines show?

```python
print(card["team"])
print(card["number"])
```

`card["team"]` means: "from `card`, give me the value with the label `team`".
It's the same square brackets as `lineup[0]` in Lab 4 and `stats["HR"]` in Lab 2:
brackets after a name **pick something out of it**. For a list you give a spot number, for a dictionary you give a key.

What if you ask for a label that isn't on the card? Run this box. It shows an error **on purpose**:

```python
print(card["age"])
```

`KeyError: 'age'`. You met `KeyError` in Lab 2, when you asked the table for a column it didn't have.
Same thing here: there's no key called `"age"` on this card. And like always, the spelling must be exact: `"Team"` wouldn't work either.

### Adding and changing facts
A card can grow. **Guess first:** how many facts will the card have after this box?

```python
card = {
    "name": "Aaron Judge",
    "team": "New York Yankees",
    "number": 99,
    "position": "right field",
}

card["best season HR"] = 62
card["position"] = "center field"

print(card)
print("Facts on the card:", len(card))
```

**5** facts. Both new lines look the same, but they did different jobs:

- `card["best season HR"] = 62`: there was **no** such key, so Python **added** a new pair. (Judge hit 62 home runs in 2022.)
- `card["position"] = "center field"`: that key was **already there**, so Python **changed** its value.
  (Judge has played center field too.) A key can only be on the card once.

It's your old friend `=`: "put into the box". Here the box is one spot on the card.

## Part 2. The scorekeeper's secret code 🔢
Baseball scorekeepers don't write "shortstop". They write a **number**. Every position on the field has one, 1 to 9.
A dictionary is perfect for a code like that: the key is the number, the value is the position.

**Guess first:** what is position 6?

```python
positions = {
    1: "pitcher",
    2: "catcher",
    3: "first baseman",
    4: "second baseman",
    5: "third baseman",
    6: "shortstop",
    7: "left fielder",
    8: "center fielder",
    9: "right fielder",
}

print(positions[6])
```

**Shortstop.** Keys don't have to be text: here they're numbers, so no quotes.

**Careful, this is not a list!** In a list, `[6]` means "spot 6, counting from 0": the **7th** thing.
In this dictionary, `6` is just a **label** we chose, so `positions[6]` gives exactly what's written next to `6`.
There's no counting from 0, and no key `0` at all.

When you hear "a **6-4-3 double play**", that's this code: the ball went from player 6, to player 4, to player 3.
A loop from Lab 5 can read it out:

```python
play = [6, 4, 3]

for number in play:
    print(number, positions[number])
```

Shortstop to second baseman to first baseman: two outs on one ball.

- `play` is a **list** of numbers.
- The loop puts each number into the box `number`.
- `positions[number]` looks that number up in the **dictionary**.

### Is that key in there?
You know `in` from Lab 4. With a dictionary, `in` asks: "is this **key** in there?"
Here it is inside a machine, like the ones you built in Lab 6. **Guess first:** what will the two lines show?

```python
def position_name(number):
    if number in positions:
        return positions[number]
    else:
        return "no such position"

print(position_name(2))
print(position_name(12))
```

`catcher`, then `no such position`. Without the `if`, `positions[12]` would stop everything with a `KeyError`.
Asking `in` first keeps your machine safe.

## Part 3. Reading a whole card 📖
`print(card)` squeezes everything onto one line. A loop can show the card nicely, one fact per line.
**Guess first:** when a loop walks through a dictionary, what goes into the box each time: the key or the value?

```python
card = {
    "name": "Aaron Judge",
    "team": "New York Yankees",
    "number": 99,
    "position": "right field",
}

for key in card:
    print(key, ":", card[key])
```

A loop over a dictionary gives you the **keys**, one at a time: `"name"`, then `"team"`, and so on.
With the key in hand, `card[key]` gets its value. (No quotes around `key` here: it's a box, not the word "key". Lab 2!)

### A team of cards
One card is one dictionary. A **team** is a **list of dictionaries**. This time each card is on one line, to save space:

```python
cards = [
    {"name": "Aaron Judge", "team": "New York Yankees", "number": 99},
    {"name": "Shohei Ohtani", "team": "Los Angeles Dodgers", "number": 17},
    {"name": "Cal Raleigh", "team": "Seattle Mariners", "number": 29},
]

print(cards[1])
print(cards[1]["name"])
```

- `cards[1]["name"]` goes one step further. Read it left to right: from `cards`, take spot 1; from that card, take `"name"`.

## Part 4. Cards made from real stats 📊
Card companies don't type every number by hand. Let's build a machine that makes a card from the real data.
First load the data with the hidden box (click ▶, about 20 seconds). Remember: one row is one player's whole season.

```python
#@title ⚾ Load the MLB data, 1871 to today (click ▶ and wait about 20 seconds)
import datetime
import unicodedata

import pandas as pd
import requests

LAHMAN = "https://raw.githubusercontent.com/daviddalpiaz/pylahman/main/data-raw/"
MLB = "https://statsapi.mlb.com/api/v1/"
STATS = ["G", "AB", "H", "HR", "RBI", "SB", "BB", "SO"]


def plain(text):
    # "Acuña" -> "Acuna": kids type names on an English keyboard
    return unicodedata.normalize("NFKD", str(text)).encode("ascii", "ignore").decode()


def short(text):
    # "Acuña Jr." -> "acu": the start of a last name, to match players across the two sources
    return plain(text).lower()[:3]


# History: the Lahman Baseball Database (SABR, CC BY-SA 3.0). One row per player, season and team.
batting = pd.read_csv(LAHMAN + "Batting.csv")
people = pd.read_csv(LAHMAN + "People.csv")
teams = pd.read_csv(LAHMAN + "Teams.csv")[["yearID", "teamID", "name"]].rename(columns={"name": "team"})
batting = batting.merge(teams, on=["yearID", "teamID"], how="left")

# A player traded during a season has a row for each team: add them up and keep his last team.
per_season = batting.groupby(["playerID", "yearID"], sort=False)
history = per_season[STATS].sum(min_count=1).join(per_season["team"].last()).reset_index()
people["name"] = people["nameFirst"] + " " + people["nameLast"]
history = history.merge(people[["playerID", "name"]], on="playerID").rename(columns={"yearID": "season"})

# The seasons the history book doesn't have yet come live from MLB (statsapi.mlb.com).
lahman_ids = {
    (int(p.birthYear), int(p.birthMonth), int(p.birthDay), short(p.nameLast)): p.playerID
    for p in people.dropna(subset=["birthYear", "birthMonth", "birthDay"]).itertuples()
}
mlb_names, live = {}, []
try:
    for season in range(int(history["season"].max()) + 1, datetime.date.today().year + 1):
        url = MLB + f"stats?stats=season&group=hitting&season={season}&sportId=1&playerPool=ALL&limit=5000"
        found = requests.get(url, timeout=60).json().get("stats", [])
        hitters = found[0]["splits"] if found else []
        if not hitters:
            continue
        roster = requests.get(MLB + f"sports/1/players?season={season}", timeout=60).json()["people"]
        born = {p["id"]: p.get("birthDate", "") for p in roster}
        for h in hitters:
            p, s = h["player"], h["stat"]
            y, m, d = (born.get(p["id"]) or "0-0-0").split("-")
            player_id = lahman_ids.get((int(y), int(m), int(d), short(p["lastName"])), f"mlb{p['id']}")
            mlb_names[player_id] = p["fullName"]
            live.append({
                "playerID": player_id, "season": season, "team": h.get("team", {}).get("name"), "name": p["fullName"],
                "G": s["gamesPlayed"], "AB": s["atBats"], "H": s["hits"], "HR": s["homeRuns"], "RBI": s["rbi"],
                "SB": s["stolenBases"], "BB": s["baseOnBalls"], "SO": s["strikeOuts"],
            })
except (requests.RequestException, ValueError, KeyError) as problem:
    live = []
    print("MLB's live stats didn't answer, so the newest season is missing today. Everything else works.", problem)

stats = pd.concat([history, pd.DataFrame(live)], ignore_index=True)
# Players active today get the name MLB uses now on their whole career ("Vladimir Guerrero Jr."), without accents.
stats["name"] = stats["playerID"].map(mlb_names).fillna(stats["name"]).map(plain)
# Whole numbers (114, not 114.0); a stat nobody counted back then stays empty: <NA>.
stats[STATS] = stats[STATS].astype("Int64")
stats = stats.sort_values(["playerID", "season"], ignore_index=True)
stats = stats[["name", "season", "team"] + STATS + ["playerID"]]
print("Ready!", len(stats), "rows, seasons", stats["season"].min(), "to", stats["season"].max())
```

This machine starts with an **empty** card, `{}`, and adds the facts one by one, the way you added `"best season HR"` in Part 1.
The right sides are all things you know: `len( )`, `.sum()` and `.max()` from Lab 2.

```python
def make_card(player):
    seasons = stats[stats["name"] == player]
    card = {}
    card["name"] = player
    card["seasons"] = len(seasons)
    card["home runs"] = seasons["HR"].sum()
    card["best season"] = seasons["HR"].max()
    return card

judge = make_card("Aaron Judge")
print(judge["home runs"])
```

| line | what it does |
|---|---|
| `seasons = stats[stats["name"] == player]` | keep only this player's rows |
| `card = {}` | an empty card: curly brackets with nothing inside |
| `card["name"] = player` | add the first pair. The value is whatever is in the box `player`. |
| `card["seasons"] = len(seasons)` | how many rows = how many seasons |
| `card["home runs"] = seasons["HR"].sum()` | add up all his home runs |
| `card["best season"] = seasons["HR"].max()` | the most home runs in one season |
| `return card` | hand back the whole finished card |

A machine can `return` a whole dictionary: four answers in one box. Judge's card says **386** home runs (through 2026).

Now a second machine that **shows** any card nicely. It's your loop from Part 3 with a `def` around it:

```python
def show_card(card):
    print("🃏")
    for key in card:
        print(key, ":", card[key])

show_card(judge)
```

And both machines together, in a loop over a team. **Guess first:** who has the most home runs on his card?

```python
my_team = ["Aaron Judge", "Shohei Ohtani", "Hank Aaron"]

for player in my_team:
    show_card(make_card(player))
```

Hank Aaron: **755** home runs in 23 seasons.
`show_card(make_card(player))` is read from the inside out, like in Lab 2: first make the card, then show it.
