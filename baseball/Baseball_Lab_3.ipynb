## Part 1. Yes or no? 🤔
Python can answer yes-or-no questions. It answers with **`True`** (yes) or **`False`** (no).

**Guess first:** what will these two lines show?

````python
print(10 > 7)
print(3 > 7)
````

- `10 > 7` asks: "is 10 **bigger than** 7?" Yes → `True`.
- `3 > 7` asks: "is 3 bigger than 7?" No → `False`.

`True` and `False` are special Python words: they start with a capital letter and have **no quotes**.

Here are the question symbols:

| symbol | the question it asks |
|---|---|
| `>` | is it bigger than? |
| `<` | is it smaller than? |
| `>=` | is it **at least**? (bigger, or the same) |
| `<=` | is it **at most**? (smaller, or the same) |
| `==` | is it the same? (remember from Lab 2: two `=`) |

A batter gets a walk after **4 balls**. **Guess first:** the pitcher just threw ball number 4.
Which line says `True`?

````python
balls_for_a_walk = 4
balls = 4

print(balls > balls_for_a_walk)
print(balls >= balls_for_a_walk)
````

`4 > 4` is `False`: 4 is not **bigger** than 4.
`4 >= 4` is `True`: 4 is **at least** 4. That's the question the umpire really asks, and so will we.

## Part 2. `if`: the first-base umpire ⚾
A runner races to first base, and the fielder throws the ball there. Whoever gets there first wins.
Read the code like a sentence:
**"if the runner's time is smaller than the throw's time, say Safe!"**

````python
runner_time = 4.1   # seconds from home plate to first base
throw_time = 4.3    # seconds until the ball is in the first baseman's glove

if runner_time < throw_time:
    print("Safe! 🙌")
````

Three new things:
- **`if`** starts a question. If the answer is `True`, Python does what's underneath.
  If it's `False`, Python skips it.
- The **colon** `:` at the end of the `if` line means "here's what to do".
- The line underneath is pushed to the right with **4 spaces**. That push is called an **indent**,
  and it shows Python which lines belong to the `if`. In Colab, press Enter after the colon
  and it adds the spaces for you.

What if we forget the colon? Run this box. It shows an error **on purpose**:

````python
if runner_time < throw_time
    print("Safe! 🙌")
````

The last line says `SyntaxError: expected ':'`. Python is telling you exactly what's missing: the colon.

### `else`: what to do otherwise
`else` means **"otherwise"**. It runs only when the `if` question was `False`.
Look: `else` has a colon too, and its line is indented too.

````python
runner_time = 4.5
throw_time = 4.3

if runner_time < throw_time:
    print("Safe! 🙌")
else:
    print("Out! ✊")
````

## Part 3. `elif`: how far did it go? 🏟️
The batter hits a fly ball. Our ballpark has **three** possible calls.
Two choices need `if` and `else`. For more, we use **`elif`**, short for "else if": *otherwise, if...*

Our ballpark's rules (the outfield wall is 400 feet away):
- 400 feet or more → **home run**
- otherwise, 300 feet or more → it flies over the outfielders' heads: **extra bases**
- otherwise → an outfielder **catches it**

````python
distance = 250   # feet

if distance >= 400:
    print("Home run! 💣")
elif distance >= 300:
    print("Over their heads! Extra bases 🏃")
else:
    print("Caught! The batter is out 🧤")
````

How Python reads it: it asks the questions **from top to bottom**. The **first** `True` wins,
and Python skips everything below it. If nothing is `True`, it does the `else`.

(Real fly balls also depend on the wind, the wall's height and how fast the outfielders are. Our bot keeps it simple.)

## Part 4. Rate your player's best season ⭐
Let's use `if` on real stats. First load the data with the hidden box, the same way as in Lab 2 (click ▶, about 20 seconds):

````python
#@title ⚾ Load the MLB data, 1871 to today (click ▶ and wait about 20 seconds)
import datetime
import unicodedata

import pandas as pd
import requests

LAHMAN = "https://raw.githubusercontent.com/daviddalpiaz/pylahman/main/data-raw/"
MLB = "https://statsapi.mlb.com/api/v1/"
STATS = ["G", "AB", "H", "HR", "RBI", "SB", "BB", "SO"]


def plain(text):
    # "Acuña" -> "Acuna": kids type names on an English keyboard
    return unicodedata.normalize("NFKD", str(text)).encode("ascii", "ignore").decode()


def short(text):
    # "Acuña Jr." -> "acu": the start of a last name, to match players across the two sources
    return plain(text).lower()[:3]


# History: the Lahman Baseball Database (SABR, CC BY-SA 3.0). One row per player, season and team.
batting = pd.read_csv(LAHMAN + "Batting.csv")
people = pd.read_csv(LAHMAN + "People.csv")
teams = pd.read_csv(LAHMAN + "Teams.csv")[["yearID", "teamID", "name"]].rename(columns={"name": "team"})
batting = batting.merge(teams, on=["yearID", "teamID"], how="left")

# A player traded during a season has a row for each team: add them up and keep his last team.
per_season = batting.groupby(["playerID", "yearID"], sort=False)
history = per_season[STATS].sum(min_count=1).join(per_season["team"].last()).reset_index()
people["name"] = people["nameFirst"] + " " + people["nameLast"]
history = history.merge(people[["playerID", "name"]], on="playerID").rename(columns={"yearID": "season"})

# The seasons the history book doesn't have yet come live from MLB (statsapi.mlb.com).
lahman_ids = {
    (int(p.birthYear), int(p.birthMonth), int(p.birthDay), short(p.nameLast)): p.playerID
    for p in people.dropna(subset=["birthYear", "birthMonth", "birthDay"]).itertuples()
}
mlb_names, live = {}, []
try:
    for season in range(int(history["season"].max()) + 1, datetime.date.today().year + 1):
        url = MLB + f"stats?stats=season&group=hitting&season={season}&sportId=1&playerPool=ALL&limit=5000"
        found = requests.get(url, timeout=60).json().get("stats", [])
        hitters = found[0]["splits"] if found else []
        if not hitters:
            continue
        roster = requests.get(MLB + f"sports/1/players?season={season}", timeout=60).json()["people"]
        born = {p["id"]: p.get("birthDate", "") for p in roster}
        for h in hitters:
            p, s = h["player"], h["stat"]
            y, m, d = (born.get(p["id"]) or "0-0-0").split("-")
            player_id = lahman_ids.get((int(y), int(m), int(d), short(p["lastName"])), f"mlb{p['id']}")
            mlb_names[player_id] = p["fullName"]
            live.append({
                "playerID": player_id, "season": season, "team": h.get("team", {}).get("name"), "name": p["fullName"],
                "G": s["gamesPlayed"], "AB": s["atBats"], "H": s["hits"], "HR": s["homeRuns"], "RBI": s["rbi"],
                "SB": s["stolenBases"], "BB": s["baseOnBalls"], "SO": s["strikeOuts"],
            })
except (requests.RequestException, ValueError, KeyError) as problem:
    live = []
    print("MLB's live stats didn't answer, so the newest season is missing today. Everything else works.", problem)

stats = pd.concat([history, pd.DataFrame(live)], ignore_index=True)
# Players active today get the name MLB uses now on their whole career ("Vladimir Guerrero Jr."), without accents.
stats["name"] = stats["playerID"].map(mlb_names).fillna(stats["name"]).map(plain)
# Whole numbers (114, not 114.0); a stat nobody counted back then stays empty: <NA>.
stats[STATS] = stats[STATS].astype("Int64")
stats = stats.sort_values(["playerID", "season"], ignore_index=True)
stats = stats[["name", "season", "team"] + STATS + ["playerID"]]
print("Ready!", len(stats), "rows, seasons", stats["season"].min(), "to", stats["season"].max())
````

This box finds a player's best season, like in Lab 2, and then **rates** it:
50 home runs or more is a monster season, 30 or more is a big power season.

````python
name = "Aaron Judge"

seasons = stats[stats["name"] == name]
best = seasons["HR"].max()
print(name, "best season:", best, "home runs")

if best >= 50:
    print("MONSTER SEASON! 🔥")
elif best >= 30:
    print("Big power season! 💥")
else:
    print("Solid, but no 30-homer season yet.")
````

## Part 5. Inside the `compare` machine 🔧
Remember `compare` from Lab 2? Now you can read it. Here it is, line by line:

| line | what it means |
|---|---|
| `def compare(player1, player2, stat):` | "teach Python a new command called `compare`; it needs two players and a stat" |
| `total1 = ...` and `total2 = ...` | add up each player's whole career in that stat |
| `if total1 > total2:` | is player 1's total bigger? |
| `else:` | otherwise... |

Run it, and then try the test below it.

````python
def compare(player1, player2, stat):
    total1 = stats[stats["name"] == player1][stat].sum()
    total2 = stats[stats["name"] == player2][stat].sum()
    print(player1, ":", total1)
    print(player2, ":", total2)
    if total1 > total2:
        print("Winner:", player1, "🏆")
    else:
        print("Winner:", player2, "🏆")
````

**The test:** compare Aaron Judge with... Aaron Judge. Their numbers are exactly the same.
What **should** the machine say? Run it and see what it **does** say.

````python
compare("Aaron Judge", "Aaron Judge", "HR")
````

🐞 **You found a bug!** When the totals are the same, `total1 > total2` is `False`,
so the machine goes to `else` and picks player 2 as the winner. But it's a tie!
