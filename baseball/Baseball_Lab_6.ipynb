## Part 1. Your first machine 🔧
**Guess first:** what will this box show?

```python
def cheer():
    print("Let's go, Yankees!")
    print("👏 👏 👏")
```

**Nothing!** That's not a mistake. `def` is short for **define**: it **teaches** Python a new command called `cheer`.
Python learned it, and now waits for you to **use** it. Building a machine and switching it on are two different things.

To switch it on, write its name with round brackets. **Guess first:** how many lines will show up?

```python
cheer()
cheer()
```

Four lines: the machine ran **twice**, and each run does both `print` lines.

| piece | what it does |
|---|---|
| `def` | "I'm teaching you a new command" |
| `cheer` | the command's name. You pick it. |
| `( )` | round brackets: every machine has them, like `print( )` and `len( )` |
| `:` | the colon, like after `if` and `for`: "here comes what the machine does" |
| the pushed-in lines | the **inside** of the machine. Pushed in 4 spaces. |
| `cheer()` | **use** the machine. People say "call" it. |

What if you forget to push the inside in? Run this box. It shows an error **on purpose**:

```python
def cheer():
print("Let's go, Yankees!")
```

`IndentationError: expected an indented block after function definition`.
It's the same error as with `if` (Lab 3) and `for` (Lab 5), for the same reason:
the pushed-in lines are how Python knows what's **inside** the machine. No pushed-in lines, an empty machine.

### A machine with an input
This machine only cheers for the Yankees. Let's make one that cheers for **any** team.
**Guess first:** what will the two lines at the bottom show?

```python
def cheer(team):
    print("Let's go,", team, "!")

cheer("Dodgers")
cheer("Mariners")
```

`team` inside the round brackets is an **input**: an empty box that gets filled each time you use the machine.
(Python people call it a **parameter**.)

- `cheer("Dodgers")` puts `"Dodgers"` into the box `team`, then runs the inside.
- `cheer("Mariners")` puts `"Mariners"` into `team`, and runs the inside again.

You wrote the machine **once** and used it for two teams. You've done this before with other people's machines:
in `compare("Aaron Judge", "Shohei Ohtani", "HR")` you filled three boxes.

## Part 2. `return`: a machine that hands back an answer 🎁
`cheer` and `announce` **show** something. Most machines do something more useful: they work out an answer and **hand it back** to you.

Let's build a real baseball one. A **batting average** tells you how often a batter gets a hit:
take his **hits** and share them out over his **at-bats** (his turns at bat). That's division, `/`.

In 1941, Ted Williams got **185** hits in **456** at-bats, one of the most famous seasons ever.

```python
def batting_average(hits, at_bats):
    return hits / at_bats

print(batting_average(185, 456))
```

**`return`** means: "the machine is done, and **this** is the answer it hands back".
`batting_average(185, 456)` hands back the number, and `print` shows it.

But what a long number! Baseball people use only 3 digits after the dot. You know `round( )` from Lab 2.
A machine can have more than one line inside:

```python
def batting_average(hits, at_bats):
    average = hits / at_bats
    return round(average, 3)

print(batting_average(185, 456))
```

**0.406**. Baseball people say "four-oh-six". Nobody has hit over .400 for a whole season since.

| line | what it does |
|---|---|
| `def batting_average(hits, at_bats):` | a new machine with two input boxes |
| `average = hits / at_bats` | hits divided by at-bats, put into a box called `average` |
| `return round(average, 3)` | round it to 3 digits and hand it back |

### `print` or `return`: what's the difference?
Here's almost the same machine, but it uses `print` instead of `return`.
We try to save its answer in a box. **Guess first:** what's in the box?

```python
def average_shown(hits, at_bats):
    print(round(hits / at_bats, 3))

answer = average_shown(185, 456)
print("In the answer box:", answer)
```

The machine showed 0.406 on the screen... and the box holds **`None`**. `None` is Python's word for "nothing".

- **`print`** shows something on the screen for **you** to read. The machine hands back nothing.
- **`return`** hands the answer back to **Python**, so you can put it in a box, compare it, or do math with it.

A scoreboard *shows* the score. An umpire *hands* you the ball. With `return`, you get the ball.

## Part 3. Rebuild `compare` 🏗️
Time to rebuild the `compare` machine from Lab 2, and make it better: this one will **hand back** the winner.
First load the data with the hidden box (click ▶, about 20 seconds). Remember: one row is one player's whole season.

```python
#@title ⚾ Load the MLB data, 1871 to today (click ▶ and wait about 20 seconds)
import datetime
import unicodedata

import pandas as pd
import requests

LAHMAN = "https://raw.githubusercontent.com/daviddalpiaz/pylahman/main/data-raw/"
MLB = "https://statsapi.mlb.com/api/v1/"
STATS = ["G", "AB", "H", "HR", "RBI", "SB", "BB", "SO"]


def plain(text):
    # "Acuña" -> "Acuna": kids type names on an English keyboard
    return unicodedata.normalize("NFKD", str(text)).encode("ascii", "ignore").decode()


def short(text):
    # "Acuña Jr." -> "acu": the start of a last name, to match players across the two sources
    return plain(text).lower()[:3]


# History: the Lahman Baseball Database (SABR, CC BY-SA 3.0). One row per player, season and team.
batting = pd.read_csv(LAHMAN + "Batting.csv")
people = pd.read_csv(LAHMAN + "People.csv")
teams = pd.read_csv(LAHMAN + "Teams.csv")[["yearID", "teamID", "name"]].rename(columns={"name": "team"})
batting = batting.merge(teams, on=["yearID", "teamID"], how="left")

# A player traded during a season has a row for each team: add them up and keep his last team.
per_season = batting.groupby(["playerID", "yearID"], sort=False)
history = per_season[STATS].sum(min_count=1).join(per_season["team"].last()).reset_index()
people["name"] = people["nameFirst"] + " " + people["nameLast"]
history = history.merge(people[["playerID", "name"]], on="playerID").rename(columns={"yearID": "season"})

# The seasons the history book doesn't have yet come live from MLB (statsapi.mlb.com).
lahman_ids = {
    (int(p.birthYear), int(p.birthMonth), int(p.birthDay), short(p.nameLast)): p.playerID
    for p in people.dropna(subset=["birthYear", "birthMonth", "birthDay"]).itertuples()
}
mlb_names, live = {}, []
try:
    for season in range(int(history["season"].max()) + 1, datetime.date.today().year + 1):
        url = MLB + f"stats?stats=season&group=hitting&season={season}&sportId=1&playerPool=ALL&limit=5000"
        found = requests.get(url, timeout=60).json().get("stats", [])
        hitters = found[0]["splits"] if found else []
        if not hitters:
            continue
        roster = requests.get(MLB + f"sports/1/players?season={season}", timeout=60).json()["people"]
        born = {p["id"]: p.get("birthDate", "") for p in roster}
        for h in hitters:
            p, s = h["player"], h["stat"]
            y, m, d = (born.get(p["id"]) or "0-0-0").split("-")
            player_id = lahman_ids.get((int(y), int(m), int(d), short(p["lastName"])), f"mlb{p['id']}")
            mlb_names[player_id] = p["fullName"]
            live.append({
                "playerID": player_id, "season": season, "team": h.get("team", {}).get("name"), "name": p["fullName"],
                "G": s["gamesPlayed"], "AB": s["atBats"], "H": s["hits"], "HR": s["homeRuns"], "RBI": s["rbi"],
                "SB": s["stolenBases"], "BB": s["baseOnBalls"], "SO": s["strikeOuts"],
            })
except (requests.RequestException, ValueError, KeyError) as problem:
    live = []
    print("MLB's live stats didn't answer, so the newest season is missing today. Everything else works.", problem)

stats = pd.concat([history, pd.DataFrame(live)], ignore_index=True)
# Players active today get the name MLB uses now on their whole career ("Vladimir Guerrero Jr."), without accents.
stats["name"] = stats["playerID"].map(mlb_names).fillna(stats["name"]).map(plain)
# Whole numbers (114, not 114.0); a stat nobody counted back then stays empty: <NA>.
stats[STATS] = stats[STATS].astype("Int64")
stats = stats.sort_values(["playerID", "season"], ignore_index=True)
stats = stats[["name", "season", "team"] + STATS + ["playerID"]]
print("Ready!", len(stats), "rows, seasons", stats["season"].min(), "to", stats["season"].max())
```

Start with a small machine: one player's **career total** in any stat.
You know both inside lines from Lab 2: keep only his rows, then add up one column.

```python
def career_total(player, stat):
    seasons = stats[stats["name"] == player]
    return seasons[stat].sum()

print(career_total("Aaron Judge", "HR"))
print(career_total("Hank Aaron", "HR"))
```

386 for Judge (through 2026) and 755 for Hank Aaron. One machine, any player, any stat.

(If you spell a name wrong, the machine finds no rows and hands back **0**. A star with 0 home runs? Check the spelling.)

### ⚾ Your turn: build `compare`
A machine can **use another machine**. `compare` uses `career_total` twice, then decides with `if` / `elif` / `else` (Lab 3).
It even handles a tie, the bug you fixed in Lab 3.

Two things are missing:
- which player to add up for `total2` (look at the line above it)
- the word that hands back an answer (6 letters, you just learned it)

🔎 **Google it:** `python function return value`

```python
def compare(player1, player2, stat):
    total1 = career_total(player1, stat)
    total2 = career_total(___, stat)
    print(player1, ":", total1)
    print(player2, ":", total2)
    if total1 > total2:
        ___ player1
    elif total1 == total2:
        return "Nobody, it's a tie"
    else:
        return player2

winner = compare("Aaron Judge", "Kyle Schwarber", "HR")
print("Winner:", winner, "🏆")
```

Judge **386**, Schwarber **385**: one home run apart after the 2026 season!

- This machine does both: it **prints** the two totals for you to read, and it **returns** the winner's name.
- There are three `return` lines, but only **one** of them ever runs: the one under the rule that's true.
  `return` ends the machine right there.
- `winner = compare(...)` catches the answer in a box. That's something the Lab 2 machine couldn't do.

## Part 4. Machines + loops = a scouting report 📋
Now put your machines together. A **career** batting average is all of a player's hits shared out over all of his at-bats.
You already have a machine for each piece!

**Guess first:** is Judge's career average higher or lower than his 0.331 in 2025?

```python
def career_average(player):
    hits = career_total(player, "H")
    at_bats = career_total(player, "AB")
    return batting_average(hits, at_bats)

print(career_average("Aaron Judge"))
```

**0.291**, lower: 2025 was one of his best years. Look how short `career_average` is.
It doesn't know anything about tables or division. It just asks your other two machines.

And with a loop from Lab 5, one machine can work through a whole team:

```python
my_team = ["Aaron Judge", "Shohei Ohtani", "Cal Raleigh"]

for player in my_team:
    print(player, career_average(player))
```

Each time around the loop, `player` is the next name, and the machine gets called with it. Three players, three answers.

(If a name is spelled wrong, you'll see `nan` and a pink warning. `nan` means "not a number":
Python found no hits and no at-bats, and 0 divided by 0 has no answer.)

### ⚾ Make it yours: your own machine

Make it yours:
