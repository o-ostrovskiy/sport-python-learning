## Part 1. Your first list 📋
**Guess first:** what will this box show?

```python
my_team = ["Aaron Judge", "Shohei Ohtani", "Cal Raleigh"]
print(my_team)
```

That's a **list**: one box that holds many things, **in order**.
(Python shows the names with single quotes `'Aaron Judge'`. Single or double quotes mean the same thing.)

| piece | what it does |
|---|---|
| `[` and `]` | square brackets: where the list **starts** and **ends** |
| `"Aaron Judge"` | each name is in quotes, because names are text (Lab 2) |
| `,` | a comma between players, so Python knows where one ends and the next begins |
| `my_team =` | puts the whole list into one box called `my_team` |

**Seen this before?** In Lab 2 you wrote `stats[["name", "HR"]]`.
The inside brackets `["name", "HR"]` were a list too: a list of column names!

**Why the commas?** Watch what happens without one. **Guess first:** how many players are on this team?

```python
no_comma = ["Aaron Judge" "Shohei Ohtani"]
print(no_comma)
print(len(no_comma))
```

No red error, but it's **wrong**: Python glued the two names into one player, `Aaron JudgeShohei Ohtani`,
and `len( )` (Lab 2: "how many?") says there's only **1** player. The comma is what keeps players apart.

## Part 2. Batting... 0th?! 🤯
Now a real lineup: 9 batters. A list this long doesn't fit on one line, so we write **one player per line**.
That's allowed: Python keeps reading until it finds the closing bracket `]`.
The first batter is the **leadoff** hitter.

**Guess first:** `lineup[1]` asks for one batter from the lineup. Which one will it show?

```python
lineup = [
    "Shohei Ohtani",
    "Aaron Judge",
    "Juan Soto",
    "Cal Raleigh",
    "Bobby Witt Jr.",
    "Jose Ramirez",
    "Kyle Schwarber",
    "Julio Rodriguez",
    "Pete Alonso",
]
print(lineup[1])
```

Did you guess Shohei Ohtani, the leadoff hitter? Almost everyone does. But Python says **Aaron Judge**!

**Python counts from 0, not 1.** The number in the brackets means "how many steps from the top".
Ohtani is at the very top: **0 steps**. Judge is 1 step down.

| batting order | 1st | 2nd | 3rd | 4th | ... | 9th |
|---|---|---|---|---|---|---|
| the number Python uses | `0` | `1` | `2` | `3` | ... | `8` |

`lineup[1]` means: "from `lineup`, give me the one at spot 1".
It's the same square-bracket idea as `stats["HR"]` in Lab 2: brackets after a name **pick something out of it**.

```python
print("Leadoff:", lineup[0])
print("Batting 2nd:", lineup[1])
print("Batting 3rd:", lineup[2])
```

There are 9 batters, so the spots are 0 to 8. What if we ask for spot **9**?
Run this box. It shows an error **on purpose**:

```python
print(lineup[9])
```

The last line says `IndexError: list index out of range`.
**Index** is Python's word for the spot number. "Out of range" means there's no spot 9 in this lineup:
9 batters only go up to spot 8. **The last spot is always one less than `len( )`.**

## Part 3. Trade deadline 📞
### `.append( )`: add a player to the end
Your team just traded for a new player. **Guess first:** where on the list will he go?

```python
my_team = ["Aaron Judge", "Shohei Ohtani", "Cal Raleigh"]
my_team.append("Juan Soto")
print(my_team)
print("Players on my team:", len(my_team))
```

- **`append`** means "add to the end". Soto goes to the bottom of the list, and the team grows from 3 to 4.
- **The dot** `.` you know from Lab 2 (`.max()`): it means "do this to that".
  `my_team.append("Juan Soto")` = "to `my_team`, add Juan Soto at the end".
- The `append` line doesn't show anything by itself. It changes the list quietly. `print` shows it.

### `in`: is he on my team?
**Guess first:** what will each line say?

```python
my_team = ["Aaron Judge", "Shohei Ohtani", "Cal Raleigh"]

print("Aaron Judge" in my_team)
print("Mookie Betts" in my_team)
print("aaron judge" in my_team)
```

- `in` asks a yes-or-no question: "is this **in** the list?" Like the symbols in Lab 3, it answers `True` or `False`.
- `"Aaron Judge" in my_team` → `True`. `"Mookie Betts" in my_team` → `False`, he's not on the team.
- `"aaron judge"` → `False`! To Python, a small `a` and a big `A` are different letters (Lab 2).
  `in` only says `True` when the name is spelled **exactly** the same.

### ⚾ Your turn: the trade machine
This box puts it all together with `if` and `else` from Lab 3:
**if the player is already on the team, say so; otherwise, trade for him.**

One word is missing in the `if` line: the one that asks "is it inside the list?". It's 2 letters.
And one is missing under `else`: the one that adds a player to the end. It's 6 letters.

🔎 **Google it:** `python check if item in list` and `python add item to end of list`

```python
my_team = ["Aaron Judge", "Shohei Ohtani", "Cal Raleigh"]
player = "Mookie Betts"   # change to any player

if player ___ my_team:
    print(player, "is already on my team ✅")
else:
    my_team.___(player)
    print("Traded for", player, "📞")

print(my_team)
```

- Under `else` there are **two** indented lines. Both belong to `else`, so both happen only when the player is **not** on the team.
- The last `print(my_team)` is **not** indented, so it runs every time.

Run it twice: once with `player = "Mookie Betts"`, then with `player = "Aaron Judge"`.

## Part 4. The real top 5 📊
Real teams build their rosters from stats. Let's do it too. First load the data with the hidden box,
the same way as in Lab 3 (click ▶, about 20 seconds). Remember: one row is one player's whole season.

```python
#@title ⚾ Load the MLB data, 1871 to today (click ▶ and wait about 20 seconds)
import datetime
import unicodedata

import pandas as pd
import requests

LAHMAN = "https://raw.githubusercontent.com/daviddalpiaz/pylahman/main/data-raw/"
MLB = "https://statsapi.mlb.com/api/v1/"
STATS = ["G", "AB", "H", "HR", "RBI", "SB", "BB", "SO"]


def plain(text):
    # "Acuña" -> "Acuna": kids type names on an English keyboard
    return unicodedata.normalize("NFKD", str(text)).encode("ascii", "ignore").decode()


def short(text):
    # "Acuña Jr." -> "acu": the start of a last name, to match players across the two sources
    return plain(text).lower()[:3]


# History: the Lahman Baseball Database (SABR, CC BY-SA 3.0). One row per player, season and team.
batting = pd.read_csv(LAHMAN + "Batting.csv")
people = pd.read_csv(LAHMAN + "People.csv")
teams = pd.read_csv(LAHMAN + "Teams.csv")[["yearID", "teamID", "name"]].rename(columns={"name": "team"})
batting = batting.merge(teams, on=["yearID", "teamID"], how="left")

# A player traded during a season has a row for each team: add them up and keep his last team.
per_season = batting.groupby(["playerID", "yearID"], sort=False)
history = per_season[STATS].sum(min_count=1).join(per_season["team"].last()).reset_index()
people["name"] = people["nameFirst"] + " " + people["nameLast"]
history = history.merge(people[["playerID", "name"]], on="playerID").rename(columns={"yearID": "season"})

# The seasons the history book doesn't have yet come live from MLB (statsapi.mlb.com).
lahman_ids = {
    (int(p.birthYear), int(p.birthMonth), int(p.birthDay), short(p.nameLast)): p.playerID
    for p in people.dropna(subset=["birthYear", "birthMonth", "birthDay"]).itertuples()
}
mlb_names, live = {}, []
try:
    for season in range(int(history["season"].max()) + 1, datetime.date.today().year + 1):
        url = MLB + f"stats?stats=season&group=hitting&season={season}&sportId=1&playerPool=ALL&limit=5000"
        found = requests.get(url, timeout=60).json().get("stats", [])
        hitters = found[0]["splits"] if found else []
        if not hitters:
            continue
        roster = requests.get(MLB + f"sports/1/players?season={season}", timeout=60).json()["people"]
        born = {p["id"]: p.get("birthDate", "") for p in roster}
        for h in hitters:
            p, s = h["player"], h["stat"]
            y, m, d = (born.get(p["id"]) or "0-0-0").split("-")
            player_id = lahman_ids.get((int(y), int(m), int(d), short(p["lastName"])), f"mlb{p['id']}")
            mlb_names[player_id] = p["fullName"]
            live.append({
                "playerID": player_id, "season": season, "team": h.get("team", {}).get("name"), "name": p["fullName"],
                "G": s["gamesPlayed"], "AB": s["atBats"], "H": s["hits"], "HR": s["homeRuns"], "RBI": s["rbi"],
                "SB": s["stolenBases"], "BB": s["baseOnBalls"], "SO": s["strikeOuts"],
            })
except (requests.RequestException, ValueError, KeyError) as problem:
    live = []
    print("MLB's live stats didn't answer, so the newest season is missing today. Everything else works.", problem)

stats = pd.concat([history, pd.DataFrame(live)], ignore_index=True)
# Players active today get the name MLB uses now on their whole career ("Vladimir Guerrero Jr."), without accents.
stats["name"] = stats["playerID"].map(mlb_names).fillna(stats["name"]).map(plain)
# Whole numbers (114, not 114.0); a stat nobody counted back then stays empty: <NA>.
stats[STATS] = stats[STATS].astype("Int64")
stats = stats.sort_values(["playerID", "season"], ignore_index=True)
stats = stats[["name", "season", "team"] + STATS + ["playerID"]]
print("Ready!", len(stats), "rows, seasons", stats["season"].min(), "to", stats["season"].max())
```

Now find the newest season, and the 5 players who hit the most home runs in it.
Every line is something you know from Lab 2:

```python
newest = stats["season"].max()
last = stats[stats["season"] == newest]
best = last.sort_values("HR", ascending=False)
names = best["name"].head(5)
print(newest)
print(names)
```

| line | what it does |
|---|---|
| `newest = stats["season"].max()` | the biggest number in the `season` column: the newest season |
| `last = stats[stats["season"] == newest]` | keep only the rows from that season |
| `best = last.sort_values("HR", ascending=False)` | put them in order, **most** home runs first |
| `names = best["name"].head(5)` | take the name column, just the top 5 |

The numbers on the left are the table's row labels, and the last line (`Name: ..., dtype: str`) is the table's
note about the column. Ignore them both.

`names` is still a piece of the table, not a list. One new command turns it into a real list: **`list( )`**.

```python
top5 = list(names)
print(top5)
```

Square brackets, commas, quotes: it's a list now!

Now `top5` is a list, so everything from today works on it. **Guess first:**
is Aaron Judge in the top 5? (The answer may surprise you.)

```python
print("Number 1:", top5[0])
print("Is Aaron Judge in the top 5?", "Aaron Judge" in top5)
```

In 2026, Kyle Schwarber and Pete Crow-Armstrong both hit **45** home runs: a tie for number 1!
Python just puts one of them first.
